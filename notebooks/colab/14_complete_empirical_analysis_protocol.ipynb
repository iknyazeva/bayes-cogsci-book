```python
# Capstone Exemplar Pipeline: End-to-End Verification
rng = np.random.default_rng(2026)
N_trials = 100
x = rng.normal(0, 1, size=N_trials)
y = 2.5 + 1.8 * x + rng.normal(0, 0.8, size=N_trials)

with pm.Model() as capstone_exemplar:
    alpha = pm.Normal("alpha", mu=0, sigma=5)
    beta = pm.Normal("beta", mu=0, sigma=5)
    sigma = pm.HalfNormal("sigma", sigma=2)
    mu = alpha + beta * x
    y_obs = pm.Normal("y", mu=mu, sigma=sigma, observed=y)
    
    idata = pm.sample(draws=500, tune=500, chains=2, random_seed=2026, return_inferencedata=True)

summary = az.summary(idata)[["mean", "sd", "hdi_3%", "hdi_97%", "r_hat"]]
print("Capstone Exemplar Diagnostic Gate:")
print(summary)
```
