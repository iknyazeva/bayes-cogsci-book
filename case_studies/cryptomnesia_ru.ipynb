# Криптомнезия: атрибуция источника слов, которых не было. От приоров к апостериорному

Иерархическая байесовская логистическая регрессия: 159 участников, 8350 ответов, четыре эксперимента. Данные предоставлены В. А. Гершкович (СПбГУ).

## Исследовательский вопрос

**Криптомнезия** — непреднамеренное присвоение чужой идеи: идея воспринимается как знакомая, а её источник не вспоминается. Предметом анализа является механизм такой ошибки в модифицированной парадигме DRM. Главный вопрос формулируется так:

> **Кому участник припишет слово, которого на самом деле никто не произносил: себе или партнёру?**

Эта формулировка относится к группам, в которых слово на этапе запоминания не звучит вообще (чистая иллюзия). В остальных группах слово произносит партнёр. Тогда существует правильный ответ («партнёр»), и вопрос состоит в том, будет ли слово тем не менее приписано себе. Этот вариант ближе к плагиату в обычном смысле: идея уже сформирована у человека, а затем высказана другим. Обоснование обоих вариантов приведено ниже.

Задача относится к области **мониторинга источника** (Source Monitoring Framework). Помнить, что событие произошло, недостаточно: необходимо помнить его источник, то есть произнёс ли слово сам участник или услышал его от другого. Исходное предположение состоит в том, что ошибки атрибуции источника возникают, когда **внутреннюю активацию смыслового понятия принимают за реальный след внешнего события**.

Помимо главного вопроса рассматриваются три уточняющих: как изменяется атрибуция при отсроченном тестировании (через неделю); как она изменяется, если слово произнёс партнёр; связана ли она с ассоциативной близостью слова к остальным словам списка.

**Статистическая постановка.** Единичный ответ малоинформативен: он может быть результатом угадывания. Поэтому анализируются доли ответов «Я» для разных типов слов и разных групп. Ответы одного участника нельзя считать независимыми (участники различаются общей склонностью отвечать «Я»), а слова одного списка сходны между собой. Это определяет выбор иерархической модели.

## Эксперимент

### Парадигма DRM

В парадигме DRM (Deese; Roediger, McDermott) участнику предъявляют список слов, ассоциативно связанных с одним словом, которое в список не входит. Например, для списка *волны, пляж, песок, вода, спокойствие, закат, лето, соль, чайка, корабль* таким словом является «море». При последующей проверке памяти значительная доля участников уверенно сообщает, что слово «море» было в списке: возникает иллюзорное воспоминание (Roediger & McDermott, 1995).

Слово, ассоциативно связанное со всеми словами списка, называется **критическим дистрактором (КД)**. В классической парадигме КД в списке не предъявляется. В данном исследовании это верно не для всех групп: в части групп КД произносит партнёр на этапе запоминания (см. раздел «Почему план 2×2»). В таблице данных и далее в тетради КД также называется *ключевым словом списка*. Остальные слова списка называются **обычными** (стимулами списка).

Используются пять ДРМ-списков по 10 слов с КД *море, гора, книга, сон, стул*. Слова упорядочены от наиболее связанного с КД к наименее связанному и всегда предъявляются в этом порядке: иллюзия возникает только при такой последовательности.

### Атрибуция источника

В классической задаче проверяется узнавание («слово было в списке?»). Здесь требуется определить источник: **кто произносил слово — участник или партнёр?** Партнёр виртуальный: изображение мужского лица и записанный мужской голос (в тексте и в данных — «Миша»).

Предполагаемый механизм состоит в следующем. При предъявлении слов списка КД активируется у участника самостоятельно, даже если его никто не произносил. Активированное слово легко извлекается из памяти; при отсутствии отчётливого следа того, что его произнёс другой, легко извлекаемое слово воспринимается как произнесённое самим участником. Ответ «Я» на КД — это **Я-атрибуция** (самоатрибуция), ошибка мониторинга источника. Это ошибка и тогда, когда КД действительно произнёс партнёр.

### Процедура

Эксперимент проводится индивидуально на компьютере. Участникам сообщают, что исследуется запоминание информации в диалоге.

**Этап 1. Запоминание: «диалог» с партнёром.** На экране — изображение партнёра и две стопки карточек. Участник выбирает стопку (клавиши «стрелка влево» и «стрелка вправо»); на 2 с появляется слово, которое необходимо произнести вслух. Через 0.5 с наступает ход партнёра: слово появляется на экране и звучит в записи мужского голоса (2 с); повторять его не требуется. Фактически предъявление не зависит от выбора стопки и следует заранее заданному порядку. Предъявляются 10 слов списка: нечётные (1, 3, 5, 7, 9) произносит участник, чётные (2, 4, 6, 8, 10) — партнёр. (Так организовано предъявление в трёх экспериментах из четырёх; в Попова_к у части участников порядок обратный, см. раздел 1.4.) После списка следует пауза 5 с, затем следующий список. Пять списков предъявляются в случайном порядке. Инструкция: запомнить слова.

**Положение КД.** Зависит от группы. В группах **без вербализации** все десять слов обычные, КД на этапе запоминания не звучит вообще. В группах **с вербализацией** десятым словом, вместо последнего обычного слова, партнёр произносит КД (в списке «море» вместо «корабль» звучит «море»). Остальные девять слов в обеих группах совпадают.

**Этап 2. Тест на атрибуцию источника.** Проводится через 5 минут или через неделю (в зависимости от группы). Слова предъявляются по одному в случайном порядке: слова списков, звучавшие на этапе запоминания, и КД всех пяти списков (КД показывают всем участникам, независимо от того, звучал ли он). На каждое слово необходимо ответить «Я» (слово произносил участник) или «Партнёр» (слово произносил партнёр). Выбор вынужденный, третьего ответа нет; слово остаётся на экране до ответа. По окончании участникам сообщают истинную цель исследования.

### Обычные слова как ориентиры

На тесте оцениваются не только КД, но и **обычные слова**, действительно звучавшие на этапе запоминания.

- Для обычного слова **правильный ответ существует**: слово произнёс участник — правильно «Я», произнёс партнёр — правильно «Партнёр». Возможны две ошибки: «Я → Партнёр» и «Партнёр → Я».
- Для КД возможны два случая. Если КД на этапе запоминания **не звучал**, правильного ответа нет: любой ответ является ошибкой, и интерес представляет то, какую ошибку выберут участники. Если КД **произнёс партнёр** (группы с вербализацией), правильный ответ — «Партнёр», а ответ «Я» — ошибка атрибуции источника.

Доля ответов «Я» на КД сама по себе неинтерпретируема. Случайному выбору не обязательно соответствует 50 %: участники различаются по общей склонности отвечать «Я» или «Партнёр» (в терминах теории обнаружения сигнала — по критерию принятия решения; помимо него различают чувствительность — способность различать «своё» и «чужое»). Поэтому КД всегда сопоставляется с обычными словами. Если КД приписывают себе чаще, чем реальные слова партнёра, эффект не сводится к общей склонности отвечать «Я». Если приписывают так же часто, как собственные слова, КД неотличим от слова, действительно произнесённого участником.

Обычные слова необходимы и для вспомогательного вопроса: способны ли участники различать источник слов и как эта способность изменяется со временем.

### Почему план 2×2

Варьируются два условия, причём межгрупповым образом: каждый участник относится ровно к одной из четырёх групп.

**1. Интервал тестирования: немедленно (через 5 минут) или через неделю (отсроченное тестирование).** Память об источнике быстро ухудшается: через неделю участник плохо помнит, кто произносил слово. Предполагается, что для КД это проявляется иначе, чем для обычных слов: сенсорные характеристики следа источника ослабевают, тогда как ощущение знакомости сохраняется, и КД приписывают себе чаще. Для проверки необходимы обе точки: немедленное тестирование как базовый уровень и отсроченное, когда информации об источнике остаётся меньше. Поскольку через неделю изменяются ответы на **все** слова, а не только на КД, изменение ответов на КД следует сопоставлять с изменением ответов на обычные слова.

**2. Звучал ли КД на этапе запоминания (вербализация).**

- *Без вербализации.* КД на этапе запоминания не звучит вообще: он появляется только на тесте. Всё, что участник о нём «помнит», он сгенерировал сам. Эти группы отвечают на главный вопрос.
- *С вербализацией.* КД произносит партнёр десятым словом списка, на месте последнего обычного слова (так список остаётся из 10 слов). У КД появляется правильный ответ — «Партнёр».

**Обоснование условия с вербализацией.** Без него были бы выявлены только иллюзия и слово, не имеющее реального источника. Условие с вербализацией введено по трём основаниям.

1. **Оно ближе к плагиату в обычном смысле.** Идея уже сформирована у человека (КД активируется самостоятельно), а затем высказывается другим. Вопрос состоит в том, будет ли слово приписано себе при наличии реального внешнего источника. С точки зрения когнитивных механизмов такое слово воспринимается как созданное самостоятельно; этим плагиат в социальном смысле отличается от процессов, происходящих в памяти.
2. **Оно позволяет проверить, защищает ли реальный след источника от самоприсвоения.** Если КД произнёс партнёр, у участника есть эпизодические детали (голос, момент предъявления). Предполагалось, что самоприсвоение при этом ослабнет, прежде всего при тестировании сразу, пока детали доступны, но не исчезнет. Через неделю детали стираются, и эффект может восстановиться. Поэтому вербализация варьируется совместно с интервалом: существенно, зависит ли эффект вербализации от времени. Результат «различий нет» также был бы информативен.
3. **Оно делает возможным сопоставление КД с реальными словами партнёра.** Приписывается ли КД чаще, чем слова, которые партнёр произнёс столь же реально? Если да, это не объясняется самим фактом произнесения партнёром.

Возможна и противоположная интерпретация: слово, произнесённое партнёром, приобретает перцептивные детали и становится ещё более доступным (насыщается эпизодическими характеристиками), поэтому приписывается себе ещё чаще. Какая из гипотез согласуется с данными, выясняется в разделе 5.

### Гипотезы и вопросы

Анализ организован вокруг четырёх вопросов. Для каждого указано, что проверяется и какой результат ожидается.

1. **Криптомнезия.** *Проверяется:* кому участники припишут КД, которого никто не произносил. КД сопоставляется с ориентирами: приписывается ли он чаще, чем реальные слова партнёра, и каково его положение относительно собственных слов участника. *Ожидание:* КД приписывают себе систематически, чаще, чем ошибочно приписывают слова партнёра. Сравнение с 0.5 для этого непригодно (см. выше).
2. **Неделя.** *Проверяется:* как изменяется присвоение КД через неделю **на фоне** общего ухудшения памяти об источнике. Поэтому сначала рассматриваются обычные слова: снижается ли точность различения «своё» и «чужое»; нет ли общего сдвига к ответу «Я» или «Партнёр». *Ожидание:* точность снижается, а присвоение КД возрастает, поскольку след источника ослабевает, а знакомость сохраняется.
3. **Вербализация.** *Проверяется:* приписывается ли КД себе, если его произнёс партнёр, и отличается ли доля присвоений от случая, когда КД не звучал; зависит ли это различие от интервала. *Ожидание:* присвоение не исчезает; предполагалось, что оно ослабевает; возможно и усиление (см. выше).
4. **Ассоциативная сила (LogDice).** *Проверяется:* приписываются ли себе чаще те слова списка, которые сильнее связаны с КД (мера LogDice). Рассматриваются обычные слова: для КД LogDice не определён. *Ожидание:* чем сильнее связь, тем выше доля ответов «Я».

Позиция слова в списке рассматривается как контрольная переменная в разделе 7.

**Операционализация.** Данные агрегируются в 12 ячеек: 6 классов слов (см. ниже) × 2 момента тестирования. Для каждой ячейки оценивается вероятность ответа «Я». Ответом на вопрос служит **разность вероятностей двух ячеек** (контраст) с интервалом неопределённости. В разделе 5 вопросам 1–3 соответствует по четыре контраста (C1–C4, D1–D4, V1–V4) плюс четыре контрольных (K1–K4); вопрос 4 рассматривается в разделе 7.

### Шесть классов слов

Слово на тесте относится к одному из шести классов. Класс определяется двумя признаками: кто произнёс слово (участник, партнёр или никто) и звучал ли КД в данной группе.

| № | Класс (имя в данных) | Описание |
|---|---|---|
| 0 | `ordinary_partner_noverb` | обычное слово, которое произнёс партнёр; КД в группе не звучал. **Опорный класс** |
| 1 | `ordinary_self_noverb` | обычное слово, которое произнёс сам участник; КД не звучал |
| 2 | `critical_unpresented` | **КД (ключевое слово), которого не произносил никто.** Чистая криптомнезия |
| 3 | `ordinary_partner_verb` | обычное слово партнёра в группе, где партнёр произнёс КД |
| 4 | `ordinary_self_verb` | обычное слово участника в той же группе |
| 5 | `critical_partner` | КД (ключевое слово), **которое действительно произнёс партнёр** |

Собственные слова в идеале признаются, слова партнёра в идеале отвергаются. Участники ошибаются, поэтому наблюдаемые доли далеки от 0 и 1; именно они задают масштаб, относительно которого оценивается КД: по расстоянию от слов партнёра (сила криптомнезии) и от собственных слов.

## Интуиция

**Криптомнезия как знакомость без контекста.** КД ощущается знакомым, поскольку к нему «подводят» все слова списка, тогда как сведения о том, кто его произнёс, в памяти не сохраняются. Если ощущение знакомости смещает ответ в сторону «Я», КД приписывают себе чаще, чем реальные слова партнёра. С течением времени контекстная информация утрачивается быстрее, чем знакомость, поэтому присвоение может возрастать.

**Наблюдаемая величина.** Для каждого класса слов рассматривается вероятность ответа «Я». Слова партнёра приписываются себе редко, собственные слова признаются часто, КД, который не звучал, занимает промежуточное положение. Расстояние от слов партнёра служит мерой силы криптомнезии. Разность двух вероятностей — **эффект на интерпретируемой шкале**: на сколько процентных пунктов КД приписывают себе чаще.

**Шкала логитов.** Вероятность принадлежит отрезку [0, 1], а линейная модель на этом отрезке выходит за его границы. Поэтому эффекты складываются на шкале логитов, где допустима вся числовая прямая, и затем переводятся в вероятности. Платой служит нелинейность: одинаковый сдвиг логита по-разному изменяет вероятность (сильнее вблизи 0.5, слабее у границ). Поэтому итоговые эффекты всегда вычисляются на шкале вероятностей.

**Числовой пример.** Пусть логит ответа «Я» на слово партнёра при тестировании сразу равен $-0.64$ (вероятность 0.35). Собственное слово добавляет $+1.30$ к логиту, КД, который не звучал, — $+0.70$. Тогда вероятности равны 0.35 для слова партнёра, 0.66 для собственного слова и 0.51 для КД: КД лежит между ориентирами. Если у другого участника логит слова партнёра равен $-1.5$ (вероятность 0.18), те же сдвиги дают 0.18, 0.45 и 0.31: сдвиги прежние, а разности вероятностей иные.

````python
# Рисунок к «Интуиции»: шкала логитов и ориентиры для ключевого слова
import matplotlib.pyplot as plt
import numpy as np
from scipy.special import expit

fig, axes = plt.subplots(1, 2, figsize=(12, 4.3))

# --- А. Логит -> вероятность: одинаковый сдвиг логита даёт разные сдвиги вероятности ---
ax = axes[0]
x = np.linspace(-5, 5, 300)
ax.plot(x, expit(x), color="0.3", lw=2)
for start, col, (tx, ty) in [(-1.5, "tab:blue", (-4.7, 0.55)), (0.0, "tab:red", (1.6, 0.35))]:
    a, b = expit(start), expit(start + 1)
    ax.plot([start, start + 1], [a, a], color=col, lw=2)
    ax.plot([start + 1, start + 1], [a, b], color=col, lw=2)
    ax.plot([start, start + 1], [a, b], "o", color=col)
    ax.annotate(f"+1 к логиту:\n{a:.2f} → {b:.2f}  (Δ = {b - a:+.2f})", (start + 1, (a + b) / 2),
                (tx, ty), fontsize=9, color=col, arrowprops=dict(arrowstyle="->", color=col))
ax.set_xlabel("логит η"); ax.set_ylabel("вероятность p = logistic(η)")
ax.set_title("А. Сдвиг логита на +1 меняет вероятность по-разному", fontsize=11)

# --- Б. Ориентиры: слово Миши, ключевое слово и своё слово (числа из примера: логиты −0.64, 0.06 и 0.66, тест сразу) ---
ax = axes[1]
vals = {"слово Миши\n(ориентир «чужое»)": expit(-0.64), "ключевое слово,\nкоторого не было": expit(0.06),
        "своё слово\n(ориентир «своё»)": expit(0.66)}
cols = ["tab:blue", "tab:red", "tab:green"]
ax.bar(range(3), list(vals.values()), color=cols, alpha=0.7)
for k, v in enumerate(vals.values()):
    ax.text(k, v + 0.015, f"{v:.2f}", ha="center")
v0, v1 = list(vals.values())[:2]
ax.annotate("", (1.45, v0), (1.45, v1), arrowprops=dict(arrowstyle="<->", color="k"))
ax.text(1.5, (v0 + v1) / 2, f"эффект\nкриптомнезии\n{v1 - v0:+.2f}", fontsize=9, va="center")
ax.axhline(0.5, color="k", ls=":", lw=1)
ax.set_xticks(range(3), list(vals.keys())); ax.set_ylim(0, 0.8); ax.set_xlim(-0.6, 2.7)
ax.set_ylabel("вероятность ответа «говорил я»")
ax.set_title("Б. Как выглядит криптомнезия\n(числа из примера ниже, тест сразу)", fontsize=11)
plt.tight_layout(); plt.show()
````

- **А. Шкала логитов.** Серая кривая — логистическая функция: она превращает логит (любое число) в вероятность (от 0 до 1). Синий и красный отрезки — один и тот же сдвиг логита на +1. Из точки с вероятностью 0.18 он поднимает вероятность на 0.20, из точки 0.50 — на 0.23. Одинаковый эффект в логитах даёт разные разности вероятностей, поэтому **итоговые эффекты вычисляются на шкале вероятностей**.

## Математическая модель: общая схема

Ответ в пробе бинарный («Я» или «Партнёр»), поэтому используется **логистическая регрессия**: вероятность ответа «Я» связывается с типом слова и моментом теста через логит. **Логит** вероятности $p$ — это $\ln\frac{p}{1-p}$, обратная функция — $\text{logistic}(\eta) = \frac{1}{1 + e^{-\eta}}$; логит 0 соответствует вероятности 0.5. В записи $N(\mu, \sigma)$ второе число — стандартное отклонение.

Модель строится последовательно, и параметры каждой модели вводятся вместе с ней:

1. **Модель A** (раздел 2.0) — только фиксированные эффекты: класс слова, момент теста и их взаимодействие. Все пробы считаются независимыми.
2. **Модель B** (раздел 4.2) — добавлены случайные сдвиги участника, списка и эксперимента: участники различаются общей склонностью отвечать «Я», списки и эксперименты — общим уровнем ответов.
3. **Модель C** (раздел 4.4) — добавлены случайные наклоны: участники различаются точностью и склонностью присваивать КД, списки — силой КД.

Каждое усложнение вводится тогда, когда проверка предыдущей модели показывает, что она не описывает данные.

### Что оценивается

Главная цель — **контрасты**: разности вероятностей ответа «Я» между ячейками «класс слова × момент теста», например «КД, который не звучал, минус слово партнёра, тест сразу». Вероятности ячеек вычисляются для типичного участника (все случайные сдвиги равны нулю) по каждому значению из апостериорного распределения; из полученной выборки контраста определяются среднее, 94 % HDI и вероятность того, что контраст положителен (раздел 5). Разбросы случайных эффектов вспомогательны: они показывают, насколько различаются участники, списки и эксперименты, и не позволяют принять различия между людьми за эффект слов.

### Главные трудности

- **Нет правильного ответа.** Для ключевого слова, которого не было, доля «я» около 0.5 сама по себе ничего не значит. Нужны ориентиры: слова Миши и свои слова. Поэтому в модели шесть классов и мы сравниваем их попарно.
- **Ответы не независимы.** Один участник отвечает 50–55 раз. Если игнорировать это, интервалы окажутся слишком узкими.
- **Схема не полностью скрещена.** Часть ячеек (например, «неделя + вербализация») измерена в одних экспериментах, часть — в других. Эффект ячейки легко перепутать с различиями между экспериментами. Всего четыре эксперимента и пять списков, поэтому разброс между ними оценивается плохо.
- **Ключевых слов мало.** На участника приходится по 5 ключевых слов, поэтому индивидуальную склонность к криптомнезии определить трудно.

Далее многократно используются одни и те же вспомогательные функции; они собраны в одном месте.

- `hdi` — **интервал наибольшей плотности (HDI)**: самый короткий интервал, который содержит 94 % выборки из апостериорного распределения. Если распределение несимметрично, он честнее обычного «хвостового» интервала.
- `post`, `draws` — достают выборку параметра из результата сэмплирования в виде обычного массива (первая ось — все $4 \times 1000$ значений цепей). Так код не зависит от версии ArviZ.
- `summarize` — таблица: среднее, стандартное отклонение, 94 % HDI и две диагностики сходимости (R-hat и ESS, о них — в разделе 4.2).
- `sample` — запускает NUTS и сообщает число *расхождений* (divergences): сигнал, что сэмплер плохо проходит область апостериорного. Заодно печатает R-hat и ESS по всем параметрам и записывает их в словарь `DIAG_LOG` (таблицы в разделах 5.6, 6.4 и 7).

````python
def hdi(x, prob=0.94):
    """Самый короткий интервал, содержащий долю prob значений выборки x."""
    x = np.sort(np.asarray(x).ravel())
    n = len(x)
    k = int(np.ceil(prob * n))
    widths = x[k - 1:] - x[: n - k + 1]
    i = int(np.argmin(widths))
    return x[i], x[i + k - 1]


def post(idata, group="posterior"):
    """Группа результата (по умолчанию апостериорное) как xarray.Dataset (работает и в ArviZ 0.x, и в 1.x)."""
    p = idata[group]
    return p.to_dataset() if hasattr(p, "to_dataset") else p


def draws(idata, name):
    """Выборка параметра: массив формы (число_значений, ...) без разделения на цепи."""
    a = post(idata)[name].transpose("chain", "draw", ...).values
    return a.reshape(-1, *a.shape[2:])


def summarize(idata, names, prob=0.94):
    """Среднее, sd, HDI и диагностики сходимости для выбранных параметров."""
    diag = az.summary(idata, var_names=list(names))
    rows = {}
    for name in names:
        ds = post(idata)[name]
        a = draws(idata, name)
        labels = [name] if a.ndim == 1 else [f"{name}[{c}]" for c in ds.coords[ds.dims[2]].values]
        a = a.reshape(a.shape[0], -1)
        for lab, col in zip(labels, a.T):
            lo, hi = hdi(col, prob)
            rows[lab] = {"mean": col.mean(), "sd": col.std(), "hdi_3%": lo, "hdi_97%": hi,
                         "ess_bulk": float(diag.loc[lab, "ess_bulk"]) if lab in diag.index else np.nan,
                         "r_hat": float(diag.loc[lab, "r_hat"]) if lab in diag.index else np.nan}
    return pd.DataFrame(rows).T


DIAG_LOG = {}                      # диагностики всех пересчётов: имя -> словарь


def sample(model, draws=DRAWS, tune=TUNE, target_accept=TARGET, log_lik=False, seed=SEED, name=None):
    """Запуск NUTS; печатает число расхождений и R-hat / ESS по всем параметрам (включая сдвиги z)."""
    with model:
        idata = pm.sample(draws=draws, tune=tune, chains=CHAINS, cores=CORES, target_accept=target_accept,
                          random_seed=seed, progressbar=False, idata_kwargs={"log_likelihood": log_lik})
    n_div = int(np.asarray(idata.sample_stats["diverging"]).sum())
    s = az.summary(idata)
    rec = {"расхождений": n_div, "из": CHAINS * draws, "R-hat (max)": float(s["r_hat"].astype(float).max()),
           "ESS bulk (min)": float(s["ess_bulk"].astype(float).min()), "ESS tail (min)": float(s["ess_tail"].astype(float).min())}
    if name is not None:
        DIAG_LOG[name] = rec
    print(f"Расхождений (divergences): {n_div} из {CHAINS * draws} значений; R-hat не больше {rec['R-hat (max)']:.3f}, "
          f"ESS bulk не меньше {rec['ESS bulk (min)']:.0f}, ESS tail не меньше {rec['ESS tail (min)']:.0f}")
    return idata
````

## 2. Модель A и приоры: что модель считает возможным до данных

### 2.0 Модель A

**Данные.** Проба $i = 1, \dots, 8350$ — один ответ участника на одно слово на тесте. Для каждой пробы известны ответ $y_i$ (1 — «Я», 0 — «Партнёр»), класс слова $k_i \in \{0, \dots, 5\}$ (таблица классов в разделе «Эксперимент»; класс 0, слово партнёра без вербализации, — опорный) и момент теста $t_i$ (0 — сразу, 1 — через неделю).

**Модель.** Ответ — испытание Бернулли с вероятностью $p_i$; логит вероятности складывается из эффекта класса слова, эффекта недели и их взаимодействия:

$$y_i \sim \text{Bernoulli}(p_i), \qquad p_i = \text{logistic}(\eta_i), \qquad \eta_i = \mu_\alpha + \beta_{k_i} + t_i\,(\beta_t + \omega_{k_i}), \qquad \beta_0 = \omega_0 = 0.$$

| Параметр | Смысл | Приор |
|---|---|---|
| $\mu_\alpha$ | логит вероятности ответа «Я» в опорном классе (слово партнёра) при тесте сразу | $N(0, 1.5)$ |
| $\beta_k$, $k = 1, \dots, 5$ | сдвиг логита класса $k$ относительно опорного при тесте сразу | $N(0, 1.5)$ |
| $\beta_t$ | сдвиг логита через неделю в опорном классе | $N(0, 1)$ |
| $\omega_k$, $k = 1, \dots, 5$ | различие эффекта недели в классе $k$ и в опорном классе (взаимодействие) | $N(0, 1)$ |

Двенадцать параметров однозначно задают вероятности 12 ячеек «класс × момент теста». Приор на $\beta_k$ шире, чем на $\beta_t$ и $\omega_k$: классы различаются сильно (на собственное слово и на слово партнёра правильные ответы противоположны). Выбор ширины приоров обоснован в разделах 2.1–2.2. Модель A предполагает, что при заданной ячейке все пробы независимы и имеют одну и ту же вероятность ответа «Я».

**Числовой пример.** Пусть $\mu_\alpha = -0.64$, $\beta_{\text{своё}} = 1.3$, $\beta_{\text{КД}} = 0.7$, $\beta_t = 0.6$, $\omega_{\text{своё}} = -1.2$, $\omega_{\text{КД}} = 0.2$ (КД — класс «КД, который не звучал»). Тогда

| Класс | $\eta$, сразу | $\eta$, через неделю | $p$, сразу / через неделю |
|---|---|---|---|
| слово партнёра | $-0.64$ | $-0.64 + 0.6 = -0.04$ | 0.35 / 0.49 |
| собственное слово | $-0.64 + 1.3 = 0.66$ | $-0.64 + 1.3 + 0.6 - 1.2 = 0.06$ | 0.66 / 0.51 |
| КД, который не звучал | $-0.64 + 0.7 = 0.06$ | $-0.64 + 0.7 + 0.6 + 0.2 = 0.86$ | 0.51 / 0.70 |

Через неделю ответы на слова партнёра и собственные слова сближаются (источник забывается), а КД приписывается себе чаще.

**Выборка из приоров (prior predictive).** Граф модели A (стрелка идёт от величины к тем, которые от неё зависят):

```
 μ_α, β_k, β_t, ω_k ─────────────┐
                                 ├──► η_i ──► p_i ──► y_i
 класс k_i, момент теста t_i ────┘
```

**Ancestral sampling** (генерация «от предков к потомкам»):
1. величины без родителей генерируются из приоров: $\mu_\alpha$, $\beta_k$, $\beta_t$, $\omega_k$;
2. вычисляются логит $\eta_i$ и вероятность $p_i$ (детерминированные функции);
3. генерируются ответы $y_i$.

Один проход даёт **один возможный сценарий эксперимента**: набор параметров и ответы, которые при них получились бы. Множество проходов задаёт **prior predictive distribution**: какие данные модель считает возможными *до наблюдения данных*. Если сценарии неправдоподобны, приоры следует пересмотреть. Для этой проверки данные не нужны.

В PyMC этот алгоритм реализован функцией `pm.sample_prior_predictive`. Модель записывается в PyMC один раз и используется как для приорной проверки, так и для оценки по данным. Ниже модель A записана в PyMC (функции `fixed_effects` и `model_A`) и из неё получено 500 сценариев. Иерархическая модель B добавляет к графу случайные сдвиги; она вводится в разделе 4.2, после того как раздел 3.4 покажет, зачем она нужна.

````python
PRIOR = dict(mu=1.5, beta=1.5, beta_t=1.0, omega=1.0, s_alpha=1.0, s_list=0.5, s_exp=0.5, s_slope=0.5)


def fixed_effects(cls, t, prior):
    """Фиксированная часть логита: μ_α + β_k + t·(β_t + ω_k). Вызывается внутри `with pm.Model():`."""
    mu_alpha = pm.Normal("mu_alpha", 0, prior["mu"])
    beta_t = pm.Normal("beta_interval", 0, prior["beta_t"])
    beta_k = pm.Normal("beta_class", 0, prior["beta"], dims="class")                # β_1..β_5
    omega_k = pm.Normal("beta_class_interval", 0, prior["omega"], dims="class")     # ω_1..ω_5
    zero = pt.zeros(1)
    bk, ok = pt.concatenate([zero, beta_k]), pt.concatenate([zero, omega_k])       # у опорного класса β_0 = ω_0 = 0
    return mu_alpha + bk[cls] + t * (beta_t + ok[cls])


def model_A(data, prior=PRIOR):
    """Модель A: только фиксированные эффекты, все пробы считаются независимыми."""
    with pm.Model(coords={"class": CLASS_ORDER[1:]}) as m:
        eta = fixed_effects(data.cls.values, data.interval.values, prior)
        pm.Bernoulli("y", logit_p=eta, observed=data.response.values)
    return m


with model_A(d):
    prior_A = pm.sample_prior_predictive(draws=500, random_seed=SEED)
print({g: list(post(prior_A, g).data_vars) for g in ["prior", "prior_predictive"]})
````

Результат содержит группу `prior` (значения параметров в 500 сценариях) и группу `prior_predictive` (смоделированные ответы $y$ для всех 8350 проб). Функция `fixed_effects` строит фиксированную часть логита и используется во всех моделях тетради; модель A далее используется в разделах 3.3 (правдоподобие) и 4.1 (оценка).

### 2.1 Свободный член $\mu_\alpha$

Базовый приор $\mu_\alpha \sim N(0, 1.5)$. Его вероятностный смысл: **логит** вероятности ответа «я» для опорного класса. Распространённое представление «неинформативный = очень широкий» здесь неверно: приор, широкий **на шкале логитов**, после перевода в вероятности становится **информативным** — он сосредотачивает почти всю массу у 0 и 1. Это главный вывод раздела, и он относится ко всем приорам модели, заданным в логитах. Сравним три ширины и вероятности, которые они задают; для отдельных распределений используется `pm.draw`.

````python
widths = [1.5, 5, 10]
p0 = expit(pm.draw(pm.Normal.dist(0, widths), draws=20_000, random_seed=SEED))      # (20000, 3): вероятность «я» в опорном классе
fig, axes = plt.subplots(1, 3, figsize=(13, 3.2), sharey=True)
for ax, s, p in zip(axes, widths, p0.T):
    ax.hist(p, np.linspace(0, 1, 41), density=True, alpha=0.75)
    ax.set_title(f"μ_α ~ N(0, {s})\nдоля p < 0.02 или p > 0.98: {np.mean((p < .02) | (p > .98)):.0%}", fontsize=10)
    ax.set_xlabel("вероятность ответа «я» в опорном классе")
plt.tight_layout(); plt.show()
````

**Вывод: приор, неинформативный на шкале логитов, информативен на шкале вероятностей.**

- При $N(0, 10)$ около 70 % значений вероятности опорного класса меньше 0.02 или больше 0.98 (при $N(0, 5)$ — 43 %). Такой приор фактически утверждает, что слово партнёра почти всегда или почти никогда получает ответ «я»: широкий на шкале логитов, он сосредотачивает массу на крайних значениях вероятности и потому сильно информативен.
- Приор $N(0, 1.5)$ даёт почти равномерное распределение вероятности на отрезке от 0.05 до 0.95; доля крайних значений около 1 %. Он не навязывает определённого значения и исключает лишь неправдоподобные вероятности порядка 0.0001. Этот приор принят как базовый.
- То же относится ко всем параметрам модели, заданным в логитах: эффектам $\beta_k$, $\beta_t$, $\omega_k$ и разбросам $\sigma$ иерархической модели (раздел 4.2). Для модели A в целом это показано в разделе 2.4.

### 2.2 Эффекты $\beta_k$, $\beta_t$, $\omega_k$ на шкале вероятностей

Приоры эффектов $N(0, 1.5)$ и $N(0, 1)$ также заданы в логитах, тогда как интерпретируется разность вероятностей. Рассмотрим, какую разность вероятностей «ключевое слово минус слово партнёра» модель считает возможной *до данных* при трёх ширинах приора на эффект (при $\mu_\alpha \sim N(0, 1.5)$).

````python
widths = [0.5, 1.5, 3.0]
mu, beta = pm.draw([pm.Normal.dist(0, 1.5), pm.Normal.dist(0, widths)], draws=50_000, random_seed=SEED)
delta = expit(mu[:, None] + beta) - expit(mu[:, None])           # приорная разность вероятностей двух ячеек: (50000, 3)
fig, axes = plt.subplots(1, 3, figsize=(13, 3.2), sharey=True)
for ax, s, dl_ in zip(axes, widths, delta.T):
    ax.hist(dl_, np.linspace(-1, 1, 61), density=True, alpha=0.75)
    ax.set_title(f"β ~ N(0, {s})", fontsize=10); ax.set_xlabel("разность вероятностей двух ячеек")
plt.tight_layout(); plt.show()
lo, hi = np.quantile(delta, [0.03, 0.97], axis=0)
pd.DataFrame({"приор на эффект": [f"N(0, {s})" for s in widths],
              "94 % приорного интервала": [f"[{a:+.2f}, {b:+.2f}]" for a, b in zip(lo, hi)],
              "P(|разность| > 0.3)": np.mean(np.abs(delta) > 0.3, axis=0).round(2),
              "P(|разность| > 0.6)": np.mean(np.abs(delta) > 0.6, axis=0).round(2)})
````

**Что видим**

- Ширина приора на эффект определяет, какие разности вероятностей модель считает возможными до данных. При $N(0, 0.5)$ почти все разности меньше 0.2 по модулю: такой приор не допускает больших эффектов.
- При базовом $N(0, 1.5)$ 94 % приорной массы лежит в $[-0.47, +0.47]$, а разность больше 0.3 по модулю имеет вероятность 0.21. Приор симметричен вокруг нуля, не исключает ни малых, ни больших эффектов и не навязывает знак.
- При $N(0, 3)$ разность больше 0.3 по модулю имеет вероятность 0.42, а больше 0.6 — 0.11: приор допускает очень большие эффекты. Для вопросов, в которых реальные различия порядка 10–30 процентных пунктов, это слишком широко.
- Базовый выбор — промежуточный. В разделе 6.1 проверим, что вывод не зависит от выбора между этими тремя.

Первый из 500 сценариев `prior_A` — один возможный исход эксперимента с настоящим планом: 159 участников, пять списков, четыре эксперимента, 8350 проб. Ниже приведены значения параметров в этом сценарии и смоделированные доли ответов «я» в 12 ячейках рядом с наблюдаемыми.

Такой исход неправдоподобен: участники почти не признают собственные слова и почти всегда присваивают чужие. Приор не содержит сведений о том, что собственные слова должны признаваться чаще слов партнёра, поскольку сдвиги классов в нём независимы. Для слабоинформативного приора это допустимо: при 200–1000 пробах в ячейке такие сценарии исключаются данными. Единичный сценарий, однако, мало говорит о приоре в целом, поэтому далее рассматривается распределение по всем сценариям.

## 3. Правдоподобие шаг за шагом

Правдоподобие отвечает на вопрос: **насколько вероятно увидеть именно такие ответы, если параметры такие-то?** Это функция параметров при фиксированных данных. Мы пойдём от одной ячейки с одним параметром к всей выборке.

### 3.1 Одна проба

Если вероятность ответа «я» в пробе равна $p$, то вероятность наблюдённого ответа — это $p$ при $y = 1$ и $1 - p$ при $y = 0$. Одной формулой: $p^{y}(1 - p)^{1 - y}$. Для 200 независимых проб с одной и той же вероятностью $p$ и $k$ ответами «я» это биномиальная вероятность $\binom{200}{k} p^k (1 - p)^{200 - k}$. Биномиальный коэффициент от $p$ не зависит, поэтому в расчётах вокруг $p$ его можно отбросить.

### 3.2 Одна ячейка, один параметр: приор × правдоподобие = апостериорное

Возьмём ячейку «ключевое слово не звучало, тест через неделю» и один параметр — вероятность $p$ ответа «я». Приор зададим на логите: $\text{logit}\, p \sim N(0, s)$. Апостериорное распределение посчитаем на сетке значений $p$ как $\text{приор} \times \text{правдоподобие}$ и нормируем. Сравним два приора: разумный $s = 1.5$ и «неосторожный» $s = 10$. И два объёма данных: все 200 проб ячейки и пять проб одного участника.

````python
p_grid = np.linspace(0.001, 0.999, 999)
dp = p_grid[1] - p_grid[0]


def grid_posterior(k, n, prior_sd):
    """Плотности приора, правдоподобия (масштабированного) и апостериорного для вероятности p на сетке."""
    prior = norm.pdf(logit(p_grid), 0, prior_sd) / (p_grid * (1 - p_grid))      # плотность p, если logit p ~ N(0, sd)
    lik = binom.pmf(k, n, p_grid)
    posterior = prior * lik
    return prior / (prior.sum() * dp), lik / (lik.sum() * dp), posterior / (posterior.sum() * dp)


cell = d[(d.stimulus_class == "critical_unpresented") & (d.interval == 1)]
per_person = cell.groupby("participant_id").response.agg(["sum", "count"])
one_id = per_person[per_person["sum"] == 4].index[0]                           # участник, ответивший «я» в 4 из 5 проб
cases = {f"вся ячейка: {int(cell.response.sum())} «я» из {len(cell)} проб": (int(cell.response.sum()), len(cell)),
         f"один участник: 4 «я» из 5 проб": (4, 5)}

fig, axes = plt.subplots(1, 2, figsize=(12, 3.8))
for ax, (title, (k, n)) in zip(axes, cases.items()):
    for sd, col in [(1.5, "tab:blue"), (10, "tab:red")]:
        prior, lik, post_ = grid_posterior(k, n, sd)
        ax.plot(p_grid, prior, "--", color=col, alpha=0.6, label=f"приор, s = {sd}")
        ax.plot(p_grid, post_, color=col, lw=2, label=f"апостериорное, s = {sd}")
    ax.plot(p_grid, lik, ":", color="k", label="правдоподобие (масштаб)")
    ax.set_title(title, fontsize=10); ax.set_xlabel("вероятность p"); ax.set_ylim(0, 8 if n == 5 else 25)
axes[0].set_ylabel("плотность"); axes[0].legend(fontsize=8, loc="upper left")
plt.tight_layout(); plt.show()
post15 = grid_posterior(4, 5, 1.5)[2]; post10 = grid_posterior(4, 5, 10)[2]
print(f"Участник, 4 «я» из 5: среднее апостериорное p = {np.sum(p_grid * post15) * dp:.2f} при s = 1.5  и  {np.sum(p_grid * post10) * dp:.2f} при s = 10")
````

**Что видим**

- **Слева (200 проб).** Оба приора дают апостериорное распределение, которое почти совпадает с правдоподобием: пик около 0.66, ширина около $\pm 0.07$. Когда данных много, приор почти не важен.
- **Справа (5 проб одного участника, 4 «я»).** Разница заметна. Разумный приор $s = 1.5$ даёт среднее 0.71 (чуть тянет к 0.5), «неосторожный» $s = 10$ — 0.80: у него много массы около 1, и он охотно верит крайним значениям. Пяти проб не хватает, чтобы вытащить апостериорное из-под влияния приора.
- Это существенно для иерархической модели. У каждого участника всего по 5 ключевых слов, и его индивидуальная склонность определяется во многом приором. В иерархической модели приор для участника берётся из распределения по остальным участникам. Это называется **частичным объединением** (partial pooling).

### 3.3 Вся выборка: сумма логарифмов, и сверка с PyMC

Для всех 8350 проб правдоподобие — **произведение** по пробам, поэтому считают его логарифм: сумму логарифмов вероятностей наблюдённых ответов. Произведение 8350 чисел, меньших единицы, компьютер округлил бы до нуля.

Используется **модель A** из раздела 2.0: только фиксированная часть, без случайных эффектов. Двенадцать параметров — по одному на каждую ячейку «класс × момент теста» — записаны через параметры $\mu_\alpha$, $\beta_k$, $\beta_t$, $\omega_k$ (раздел 2.0):

$$\eta_i = \mu_\alpha + \beta_{k[i]} + t_i\,(\beta_t + \omega_{k[i]}), \qquad y_i \sim \text{Bernoulli}(\text{logistic}(\eta_i)).$$

Логарифм правдоподобия модели A вычисляется по формуле вручную и сравнивается со значением, которое вычисляет PyMC.

````python
mA = model_A(d)

# Точка с наибольшим правдоподобием: каждая ячейка получает свою наблюдаемую долю
rate = d.groupby(["cls", "interval"]).response.mean().unstack().values           # (6, 2)
lg = logit(rate)
mu_hat, bt_hat = lg[0, 0], lg[0, 1] - lg[0, 0]
beta_hat = lg[1:, 0] - mu_hat
omega_hat = lg[1:, 1] - mu_hat - beta_hat - bt_hat
point = {"mu_alpha": mu_hat, "beta_interval": bt_hat, "beta_class": beta_hat, "beta_class_interval": omega_hat}


def loglik_manual(mu, bt, beta, omega):
    """Сумма логарифмов вероятностей наблюдённых ответов, формула из раздела «Правдоподобие»."""
    b, o = np.r_[0, beta], np.r_[0, omega]
    p = expit(mu + b[d.cls.values] + d.interval.values * (bt + o[d.cls.values]))
    y = d.response.values
    return np.sum(y * np.log(p) + (1 - y) * np.log(1 - p))


ll_manual = loglik_manual(mu_hat, bt_hat, beta_hat, omega_hat)
ll_pymc = float(mA.compile_logp(vars=[mA["y"]])(point))
print(f"log L вручную:  {ll_manual:.3f}")
print(f"log L в PyMC:   {ll_pymc:.3f}")
print(f"Для сравнения, модель с p = 0.5 в каждой пробе: {len(d) * np.log(0.5):.1f}")
````

Значения совпадают до третьего знака: PyMC считает ту же сумму логарифмов $\sum_i [\,y_i \ln p_i + (1 - y_i)\ln (1 - p_i)\,]$. Число $-5507$ — логарифм правдоподобия в точке, где каждая из 12 ячеек получила свою наблюдаемую долю (точка максимума правдоподобия). Модель с $p = 0.5$ в каждой пробе даёт $-5788$, то есть двенадцать параметров ячеек улучшают логарифм правдоподобия на 281 единицу.

Правдоподобие как функция одного параметра: возьмём $\beta$ для класса «ключевое слово не звучало» и будем менять его, остальные 11 параметров оставим как есть. Нарисуем $L(\beta) / L_{\max}$ и приор $N(0, 1.5)$.

````python
j_crit = 1                                          # beta_class[critical_unpresented]: второй элемент вектора β_1..β_5
grid = np.linspace(-0.2, 1.6, 300)
ll = np.array([loglik_manual(mu_hat, bt_hat, np.r_[beta_hat[:j_crit], b, beta_hat[j_crit + 1:]], omega_hat) for b in grid])
lik_rel = np.exp(ll - ll.max())
prior_rel = norm.pdf(grid, 0, 1.5) / norm.pdf(0, 0, 1.5)

fig, ax = plt.subplots(figsize=(7, 3.4))
ax.plot(grid, lik_rel, lw=2, label="правдоподобие $L(\\beta)/L_{\\max}$")
ax.plot(grid, prior_rel, "--", lw=2, label="приор $N(0, 1.5)$ (масштаб)")
ax.axvline(beta_hat[j_crit], color="k", ls=":", lw=1)
ax.set_xlabel("β для класса «ключевое слово: не звучало», шкала логитов"); ax.legend(fontsize=9)
plt.tight_layout(); plt.show()
lo_, hi_ = grid[lik_rel > np.exp(-1.92)][[0, -1]]                      # приближённо 95 %-й интервал по правдоподобию
print(f"Максимум правдоподобия в β = {beta_hat[j_crit]:.2f}; интервал, где log L падает не более чем на 1.92: [{lo_:.2f}, {hi_:.2f}]")
````

**Что видим.** Правдоподобие как функция $\beta$ — колокол с максимумом в $0.70$ и шириной около $\pm 0.2$. Приор $N(0, 1.5)$ на этом участке почти плоский (его высота меняется от 0.95 до 0.83 от максимума), поэтому апостериорное распределение почти совпадёт с правдоподобием: решают данные. Интервал, где логарифм правдоподобия падает не более чем на 1.92, — приближённый 95 % интервал по правдоподобию, $[0.50, 0.90]$.

### 3.4 Можно ли считать пробы независимыми?

Модель A считает, что при заданной ячейке все пробы независимы: все участники в ячейке однородны. Проверим это на данных. Если это верно, то доля ответов «я» у отдельного участника в классе должна разбрасываться только из-за случайности (биномиальный шум). Смоделируем 400 таких «независимых» выборок: каждая проба получает вероятность ответа «я», равную наблюдаемой доле её ячейки. Для каждой выборки посчитаем **стандартное отклонение долей по участникам** в каждом из шести классов и сравним с настоящим.

````python
grp = d.j.values * 6 + d.cls.values                      # номер пары «участник × класс»
order = np.argsort(grp, kind="stable")
grp_sorted = grp[order]
starts = np.r_[0, np.flatnonzero(np.diff(grp_sorted)) + 1]
cnt = np.diff(np.r_[starts, len(grp_sorted)])
cls_of_grp = grp_sorted[starts] % 6


def part_sd_by_class(y):
    """Стандартное отклонение (по участникам) доли «я» в каждом из 6 классов; y: (S, n_obs) или (n_obs,)."""
    y = np.atleast_2d(y)
    rate_pc = np.add.reduceat(y[:, order], starts, axis=1) / cnt                 # доля «я» у каждой пары участник×класс
    return np.column_stack([rate_pc[:, cls_of_grp == c].std(axis=1, ddof=1) for c in range(6)])


def sd_check_plot(y_rep, title, ax):
    """Настоящее SD по участникам (крестики) и интервал 94 % по смоделированным выборкам."""
    sd_rep = part_sd_by_class(y_rep)
    lo, hi = np.quantile(sd_rep, [0.03, 0.97], axis=0)
    ax.vlines(range(6), lo, hi, color="tab:blue", lw=8, alpha=0.35, label="94 % смоделированных выборок")
    ax.plot(range(6), sd_rep.mean(axis=0), "o", color="tab:blue", label="среднее по выборкам")
    ax.plot(range(6), sd_obs, "kx", ms=10, mew=2, label="настоящие данные")
    ax.set_xticks(range(6), CLASS_SHORT, fontsize=8); ax.set_ylabel("SD доли «я» по участникам")
    ax.set_title(title, fontsize=10)
    return (sd_obs < lo) | (sd_obs > hi)


sd_obs = part_sd_by_class(d.response.values)[0]
p_cell = expit(logit(rate)[d.cls.values, d.interval.values])                       # вероятность из ячейки для каждой пробы
y_rep_A = (rng.random((400, len(d))) < p_cell).astype(np.int8)
fig, ax = plt.subplots(figsize=(8.5, 3.8))
outside = sd_check_plot(y_rep_A, "Если пробы независимы, участники различаются так:", ax)
ax.legend(fontsize=8); plt.tight_layout(); plt.show()
print("Классы, где настоящий разброс вне интервала независимых выборок:",
      [CLASS_ORDER[c] for c in np.flatnonzero(outside)])
````

**Что видим.** Настоящее SD (крестики) выше интервала независимых выборок в трёх классах из шести: слова Миши без вербализации, свои слова без вербализации и ключевые слова, которых не было. В остальных классах крестик в верхней части интервала. Ни разу настоящее SD не оказалось ниже. Значит, участники различаются между собой сильнее, чем позволяет «независимость»: у кого-то выше склонность отвечать «я», у кого-то лучше точность. Это основание для иерархической модели. А как именно участники различаются, покажет проверка в разделе 4.

### 4.2 Модель B: три случайных сдвига

**Зачем усложнять модель.** Модель A считает все пробы независимыми при заданной ячейке. Раздел 3.4 показал, что это не так: разброс долей ответов «я» между участниками больше, чем допускает независимость. Ответы одного участника, слова одного списка и участники одного эксперимента сходны между собой; если это игнорировать, интервалы для эффектов окажутся слишком узкими.

**Модель.** К логиту модели A добавляются сдвиги участника $j[i]$, списка $m[i]$ и эксперимента $e[i]$, к которым относится проба $i$:

$$\eta_i = \underbrace{\mu_\alpha + \sigma_\alpha z_{j[i]}}_{\alpha_{j[i]}} + \underbrace{\sigma_{list}\, z_{m[i]}}_{\delta_{m[i]}} + \underbrace{\sigma_{exp}\, z_{e[i]}}_{\gamma_{e[i]}} + \beta_{k_i} + t_i\,(\beta_t + \omega_{k_i}), \qquad z \sim N(0, 1).$$

| Параметр | Смысл | Приор |
|---|---|---|
| $\alpha_j = \mu_\alpha + \sigma_\alpha z_j$ | базовый логит участника $j$ (159 участников): его общая склонность отвечать «Я»; $\mu_\alpha$ теперь — логит типичного участника | $z_j \sim N(0, 1)$ |
| $\delta_m = \sigma_{list}\, z_m$ | сдвиг списка $m$ (море, гора, книга, сон, стул) | $z_m \sim N(0, 1)$ |
| $\gamma_e = \sigma_{exp}\, z_e$ | сдвиг эксперимента $e$ (четыре эксперимента) | $z_e \sim N(0, 1)$ |
| $\sigma_\alpha$ | разброс участников на шкале логитов | $\text{HalfNormal}(1)$ |
| $\sigma_{list}$, $\sigma_{exp}$ | разброс списков и экспериментов | $\text{HalfNormal}(0.5)$ |

Остальные параметры и их приоры — как в модели A. $\text{HalfNormal}(s)$ — нормальное распределение с нулевым центром, от которого оставлена положительная половина: разброс не может быть отрицательным, и чем меньше $s$, тем сильнее приор прижимает разброс к нулю.

Сдвиги **иерархические**: у каждого участника (списка, эксперимента) свой сдвиг, но центр и разброс у них общие и оцениваются по данным. Это частичное объединение (раздел 3.2): оценка отдельного участника опирается на остальных. Запись $\mu + \sigma z$ с $z \sim N(0, 1)$ эквивалентна $N(\mu, \sigma)$ и называется *нецентрированной параметризацией*; для сэмплера она удобнее. Сдвиги **скрещены**: участник относится к одному эксперименту, но видит все пять списков, поэтому участник, список и эксперимент входят в каждую пробу независимо.

**Числовой пример.** Пусть $\mu_\alpha = -0.8$, $\sigma_\alpha = 0.15$, $\sigma_{list} = 0.05$, $\sigma_{exp} = 0.07$. Участник с $z_j = 1.2$ получает $\alpha_j = -0.8 + 0.15 \cdot 1.2 = -0.62$, список «море» с $z_m = 0.6$ — $\delta_m = 0.03$, эксперимент с $z_e = -0.7$ — $\gamma_e = -0.05$. Сумма $-0.62 + 0.03 - 0.05 = -0.64$ равна $\mu_\alpha$ из примера модели A: в модели B базовый логит складывается из общего уровня и индивидуальных сдвигов, остальная часть вычисления не меняется.

Код разбит на функции: `grouping` готовит индексы участника, списка и эксперимента, `random_shifts` добавляет сдвиги в логит, `model_B` собирает модель. Те же функции используются в моделях C и D. Как и для модели A, сначала выполняется выборка из приоров.

````python
def grouping(data):
    """Индексы участника, списка и эксперимента, пересчитанные по выбранным данным (нужно для подвыборок)."""
    j, _ = pd.factorize(data.participant_id)
    m_, _ = pd.factorize(data.list_num, sort=True)
    e, _ = pd.factorize(data.experiment)
    return dict(j=j, m=m_, e=e, n_j=int(j.max()) + 1, n_m=int(m_.max()) + 1, n_e=int(e.max()) + 1)


def random_shifts(g, prior):
    """Нецентрированные случайные сдвиги: σ·z для участника, списка и эксперимента. Вызывается внутри pm.Model."""
    sigma_alpha = pm.HalfNormal("sigma_alpha", prior["s_alpha"])
    z_alpha = pm.Normal("z_alpha", 0, 1, shape=g["n_j"])
    sigma_list = pm.HalfNormal("sigma_list", prior["s_list"])
    z_list = pm.Normal("z_list", 0, 1, shape=g["n_m"])
    shift = sigma_alpha * z_alpha[g["j"]] + sigma_list * z_list[g["m"]]
    if g["n_e"] > 1:                                  # при одном эксперименте сдвиг эксперимента не определён
        sigma_exp = pm.HalfNormal("sigma_exp", prior["s_exp"])
        z_exp = pm.Normal("z_exp", 0, 1, shape=g["n_e"])
        shift = shift + sigma_exp * z_exp[g["e"]]
    return shift


def model_B(data, prior=PRIOR):
    """Модель B: фиксированные эффекты + случайные сдвиги участника, списка, эксперимента."""
    g = grouping(data)
    with pm.Model(coords={"class": CLASS_ORDER[1:]}) as model:
        eta = fixed_effects(data.cls.values, data.interval.values, prior) + random_shifts(g, prior)
        pm.Bernoulli("y", logit_p=eta, observed=data.response.values)
    return model


with model_B(d):
    prior_B = pm.sample_prior_predictive(draws=500, random_seed=SEED)
````

Каждая пара $(\mu_\alpha, \sigma_\alpha)$ из приора задаёт **целое распределение базовой вероятности ответа «я» по участникам**. Ниже показаны 25 таких распределений из сценариев `prior_B` (159 участников в каждом).

- Слева: распределения базовой вероятности «я» у 159 участников в 25 сценариях `prior_B`. Высокие узкие пики соответствуют малому $\sigma_\alpha$ (участники почти одинаковы), растянутые и прижатые к 0 и 1 — большому (часть участников почти всегда отвечает «я», часть почти никогда).

Приоры разбросов приняты; модель B оценивается по данным.

#### Диагностика: можно ли доверять выборке?

NUTS — это не формула, а алгоритм, и он может застрять. Три проверки:

- **R-hat** сравнивает цепи между собой: если они ходят по одной и той же области, R-hat близок к 1. Допустимо до 1.01.
- **ESS** (effective sample size) — сколько независимых значений «стоит» выборка. Значения подряд в цепи похожи, поэтому ESS меньше числа значений. Нужно хотя бы несколько сотен.
- **Расхождения (divergences)** — шаги, на которых сэмплер обнаружил, что траектория расходится: это признак «узких мест» апостериорного, где сэмплер может смещать оценки.

Эти диагностики стоят в таблицах выше (`r_hat`, `ess_bulk`), а число расхождений печатает `sample`. Визуальная проверка: цепи на графике должны выглядеть как стационарные колебания около постоянного уровня и перекрываться.

````python
def chain_values(idata, label):
    """Значения параметра по цепям, форма (цепи, шаги); label вида 'beta_class[critical_unpresented]'."""
    if "[" in label:
        name, coord = label[:-1].split("[")
        da = post(idata)[name]
        da = da.sel({da.dims[2]: coord})
    else:
        da = post(idata)[label]
    return da.transpose("chain", "draw").values


def trace_plot(idata, labels):
    fig, axes = plt.subplots(len(labels), 2, figsize=(11, 1.9 * len(labels)), gridspec_kw={"width_ratios": [2.2, 1]})
    for row, lab in zip(axes, labels):
        v = chain_values(idata, lab)
        for c in range(v.shape[0]):
            row[0].plot(v[c], lw=0.5, alpha=0.8)
            row[1].hist(v[c], 40, density=True, histtype="step", lw=1.2)
        row[0].set_ylabel(lab, fontsize=8)
    axes[-1, 0].set_xlabel("шаг"); axes[-1, 1].set_xlabel("значение")
    plt.tight_layout(); plt.show()


trace_plot(idata_B, ["mu_alpha", "beta_class[critical_unpresented]", "beta_interval", "sigma_alpha"])
````

**Что видим.** Четыре цепи (четыре цвета) перемешаны и выглядят как шум вокруг постоянного уровня. Справа распределения цепей почти совпадают. Для $\sigma_\alpha$ цепь иногда «прижимается» к нулю: это нормально для параметра, ограниченного нулём, и малого по величине. Расхождений 5 из 4000 (0.1 %). То, что их мало, не позволяет их игнорировать: каждое означает, что в этом месте сэмплер мог пройти траекторию неточно. Проверим это в разделе 5.6 повторным запуском с более мелкими шагами.

### 4.3 Проверка модели B: описывает ли она данные?

Модель B даёт каждой из 12 ячеек собственный параметр. Значит, доли ячеек она воспроизведёт почти автоматически, и проверять их бессмысленно. Проверять надо то, что модель **не подгоняла напрямую**. Выберем два таких признака:

1. **Разброс между участниками.** Мы уже сравнивали настоящее SD доли «я» по участникам в каждом классе с SD независимых выборок (раздел 3.4). Теперь сравним его с выборками, смоделированными из апостериорного распределения модели B.
2. **Разброс между списками для ключевых слов.** Сильнее ли одни ключевые слова (гора, сон, книга, море, стул), чем другие? Смотрим на долю «я» для ключевых слов в каждом списке.

Для этого сделаем **posterior predictive check**: берём значения параметров из апостериорного, моделируем по ним новые ответы для тех же проб и сравниваем с настоящими.

````python
def posterior_predictive_y(model, idata, step=10, seed=SEED):
    """Ответы, смоделированные из апостериорного распределения (каждое step-е значение); форма (S, n_obs)."""
    with model:
        ppc = pm.sample_posterior_predictive(idata.sel(draw=slice(None, None, step)), var_names=["y"],
                                             random_seed=seed, progressbar=False)
    pp = ppc.posterior_predictive
    pp = pp.to_dataset() if hasattr(pp, "to_dataset") else pp
    a = pp["y"].transpose("chain", "draw", ...).values
    return a.reshape(-1, a.shape[-1]).astype(np.int8)


def group_rate_plot(y_rep, group, labels, mask, title, ax, xlabel=None):
    """Доля «я» в группах проб (внутри mask): смоделированная (интервал 94 %) и настоящая (крестик)."""
    y = d.response.values
    obs, lo, hi, mean = [], [], [], []
    for g in range(len(labels)):
        sel = (group == g) & mask
        sim = y_rep[:, sel].mean(axis=1)
        obs.append(y[sel].mean()); mean.append(sim.mean())
        l, h = np.quantile(sim, [0.03, 0.97]); lo.append(l); hi.append(h)
    obs, lo, hi = map(np.array, (obs, lo, hi))
    ax.vlines(range(len(labels)), lo, hi, color="tab:blue", lw=8, alpha=0.35, label="94 % смоделированных выборок")
    ax.plot(range(len(labels)), mean, "o", color="tab:blue", label="среднее по выборкам")
    ax.plot(range(len(labels)), obs, "kx", ms=10, mew=2, label="настоящие данные")
    ax.set_xticks(range(len(labels)), labels, fontsize=8); ax.set_title(title, fontsize=10)
    if xlabel: ax.set_xlabel(xlabel)
    return (obs < lo) | (obs > hi)


key_mask = d.cls.isin([2, 5]).values                                  # ключевые слова
y_rep_B = posterior_predictive_y(mB, idata_B)
fig, axes = plt.subplots(1, 2, figsize=(13, 3.8))
out_sd = sd_check_plot(y_rep_B, "Разброс между участниками (SD доли «я» по участникам)", axes[0])
out_list = group_rate_plot(y_rep_B, d.m.values, LIST_NAMES, key_mask, "Ключевые слова: доля «я» по спискам", axes[1])
axes[0].set_ylabel("SD доли «я» по участникам"); axes[1].set_ylabel("доля «я»")
axes[0].legend(fontsize=8); plt.tight_layout(); plt.show()
print("Модель B. Классы с настоящим SD вне интервала:", [CLASS_ORDER[c] for c in np.flatnonzero(out_sd)])
print("Модель B. Списки с настоящей долей ключевых слов вне интервала:", [LIST_NAMES[i] for i in np.flatnonzero(out_list)])
````

**Что видим**

- **Участники.** Настоящее SD (крестики) выше интервала в двух классах: свои слова без вербализации и ключевые слова, которых не было. В остальных классах оно у верхней границы. Модель B слишком «ровно» описывает людей: позволяет им отличаться только общим сдвигом. А люди различаются по классам слов: у одних хорошо различается своё и чужое слово, у других нет.
- **Списки.** Модель B предсказывает для всех пяти списков примерно одну долю присвоений ключевого слова (около 0.59), а на деле она колеблется от 0.44 (гора) до 0.74 (стул). Настоящие доли четырёх списков из пяти вне интервала. Причина: у списков в модели B только общий сдвиг, а для обычных слов списки почти не различаются, поэтому $\sigma_{list}$ получилась малой. Сила *ключевого* слова от списка к списку меняется гораздо сильнее, и модель B этого не видит.

**Вывод.** Модель B недооценивает неопределённость во всём, что касается ключевого слова. Если сила ключевого слова зависит от списка, то вывод про криптомнезию нельзя считать надёжнее, чем позволяют пять списков. Исправим это в модели C.

### 4.4 Модель C: случайные наклоны

В модели B все участники, все списки и все эксперименты отличаются от типичных только **общим сдвигом**: вверх или вниз на одно и то же число для всех классов слов. Проверка показала, что этого мало. Люди различаются не только склонностью отвечать «я» вообще, но и:

- **точностью**: насколько хорошо они различают своё слово и слово Миши (наклон по классу «своё слово»);
- **склонностью присваивать ключевое слово**: одни участники охотно, другие почти никогда (наклон по классу «ключевое слово»).

А списки различаются **силой ключевого слова**: после списка «стул» ключевое слово присваивают в 74 % случаев, после списка «гора» — в 44 %. У каждого такого различия свой разброс, и модель его оценивает.

Формально к логиту модели B добавляются три слагаемых:

$$\eta_i = \underbrace{\mu_\alpha + \beta_{k[i]} + t_i(\beta_t + \omega_{k[i]})}_{\text{эффекты}} + \underbrace{\sigma_\alpha z_{j} + \sigma_{list} z_m + \sigma_{exp} z_e}_{\text{сдвиги, как в B}} + \underbrace{\sigma_{own}\, z^{own}_j\, \text{own}_i + \sigma_{key}\, z^{key}_j\, \text{key}_i + \sigma_{list,key}\, z^{key}_m\, \text{key}_i}_{\text{случайные наклоны}}.$$

Здесь $\text{own}_i = 1$ для классов «своё слово» (1 и 4), $\text{key}_i = 1$ для ключевых слов (классы 2 и 5), в остальных пробах 0. Приоры новых разбросов $\sigma_{own}$, $\sigma_{key}$, $\sigma_{list,key} \sim \text{HalfNormal}(0.5)$: это умеренные разбросы (94 % до 0.94 в логитах), которые данные смогут увеличить, если нужно.

> *Пример.* Для типичного участника логит ключевого слова при тесте сразу был $0.06$, то есть вероятность 0.51. Участник, склонный присваивать ключевое слово ($\sigma_{key} z^{key}_j = +0.75$), получает логит $0.81$ и вероятность 0.69. Если этот же список ещё и сильный ($\sigma_{list,key} z^{key}_m = +0.66$, например, «стул»), логит $1.47$ и вероятность 0.81. В модели B это различие участников и списков по ключевому слову невозможно: у каждого одна общая добавка.

````python
def model_C(data, prior=PRIOR):
    """Модель C = B + случайные наклоны: «своё слово» и «ключевое слово» по участникам, «ключевое слово» по спискам."""
    g = grouping(data)
    own = data.cls.isin([1, 4]).values.astype(float)            # слова, которые произнёс сам участник
    key = data.cls.isin([2, 5]).values.astype(float)            # ключевые слова (звучали или нет)
    with pm.Model(coords={"class": CLASS_ORDER[1:]}) as model:
        eta = fixed_effects(data.cls.values, data.interval.values, prior) + random_shifts(g, prior)
        sigma_own = pm.HalfNormal("sigma_own", prior["s_slope"])
        z_own = pm.Normal("z_own", 0, 1, shape=g["n_j"])
        sigma_key = pm.HalfNormal("sigma_key", prior["s_slope"])
        z_key = pm.Normal("z_key", 0, 1, shape=g["n_j"])
        sigma_list_key = pm.HalfNormal("sigma_list_key", prior["s_slope"])
        z_list_key = pm.Normal("z_list_key", 0, 1, shape=g["n_m"])
        eta = (eta + sigma_own * z_own[g["j"]] * own
               + sigma_key * z_key[g["j"]] * key
               + sigma_list_key * z_list_key[g["m"]] * key)
        pm.Bernoulli("y", logit_p=eta, observed=data.response.values)
    return model


mC = model_C(d)
idata_C = sample(mC, log_lik=True, name="C")
SIGMAS_C = ["sigma_alpha", "sigma_list", "sigma_exp", "sigma_own", "sigma_key", "sigma_list_key"]
summarize(idata_C, PARAMS + SIGMAS_C).round(3)
````

**Что видим**

- **Цепи согласуются.** R-hat не больше 1.005, ESS не меньше 780; расхождение одно из 4000 (разберём в разделе 5.6).
- **Новые разбросы заметно больше нуля.** $\sigma_{own} = 0.30$ [0.20, 0.40] — участники различаются в точности. $\sigma_{key} = 0.75$ [0.48, 1.01] — ещё сильнее различаются в склонности присваивать ключевое слово. $\sigma_{list,key} = 0.66$ [0.32, 1.03] — сила ключевого слова меняется от списка к списку почти так же сильно. Для сравнения: средняя добавка к логиту за ключевое слово $\beta \approx 0.57$, то есть разброс сопоставим со средним эффектом. Общие сдвиги по-прежнему малы: $\sigma_\alpha = 0.07$, $\sigma_{list} = 0.10$, $\sigma_{exp} = 0.08$.
- **Неопределённость эффектов для ключевого слова выросла.** Для класса «ключевое слово не звучало» $\beta = 0.57 \pm 0.35$ вместо $0.68 \pm 0.16$ в модели B, для «ключевое слово сказал Миша» $\beta = 1.31 \pm 0.34$ вместо $1.32 \pm 0.17$. Оценки сдвинулись немного, а стандартные отклонения удвоились. Для обычных слов ничего не изменилось.

Доли присвоений ключевого слова по спискам внутри интервалов для всех пяти списков. Модель C воспроизводит и разброс между участниками, и различия между списками. Не всё, конечно, проверено: например, мы не смотрели на связь индивидуальных наклонов друг с другом.

#### Сравнение моделей по LOO

Графики показывают, где модель ошибается, но не дают одного числа. Для сравнения моделей используют **LOO** (leave-one-out cross-validation, быстрая оценка по PSIS): оценка того, насколько хорошо модель предсказывает **пропущенную пробу**, если все остальные ответы (в том числе тех же участников и тех же списков) известны. Показатель `elpd` — ожидаемая логарифмическая правдоподобность новой пробы, суммарная по 8350 пробам: чем он выше (ближе к нулю), тем лучше. Строки отсортированы от лучшей модели к худшей, `elpd_diff` — разница с лучшей, `dse` — её стандартная ошибка.

````python
comparison = az.compare({"A: без случайных эффектов": idata_A, "B: три сдвига": idata_B, "C: + наклоны": idata_C})
comparison
````

**Что видим.** Модель C лучше остальных на 50 единиц `elpd` при стандартной ошибке разности около 10, то есть на пять стандартных ошибок. Модели A и B практически не различаются между собой: добавление общих сдвигов участников, списков и экспериментов не улучшило предсказания, потому что эти разбросы малы. Всю пользу дало то, что добавили в C: различия между участниками и списками именно по ключевому слову и точности. Эффективное число параметров `p` выросло с 12 (A) и 39 (B) до 137 (C), и модель C всё равно выигрывает: для предсказания пропущенных проб сложность оправдана.

**Что LOO здесь не говорит.** Мы исключали по одной пробе, а остальные ответы того же участника и того же списка оставляли. Модель C выигрывает отчасти потому, что узнаёт участника и список по другим их ответам. Отсюда **не следует**, что она так же лучше предскажет *нового* участника или *новый* список: для этого нужна проверка, в которой участника или список исключают целиком (leave-one-group-out); мы её не делали. Надёжность самой оценки LOO (значения Pareto-$k$) проверена в разделе 5.6.

## 5. Эффекты: что модель говорит о криптомнезии

Дальше работаем с моделью C: она лучше предсказывает и честнее описывает неопределённость. Эффекты считаем **на шкале вероятностей**, как договорились в разделе «Интуиция».

### 5.1 Вероятности 12 ячеек

Для каждого значения параметров из апостериорного распределения считаем вероятность «я» в каждой ячейке для **типичного участника** (все случайные сдвиги и наклоны равны нулю): $p_{k,t} = \text{logistic}(\mu_\alpha + \beta_k + t(\beta_t + \omega_k))$. Получаем 4000 значений для каждой ячейки.

````python
def cell_probs(idata):
    """Вероятности «я» в 12 ячейках для типичного участника: словарь (класс, момент теста) -> выборка из апостериорного."""
    mu, bt = draws(idata, "mu_alpha"), draws(idata, "beta_interval")
    bk = np.column_stack([np.zeros(len(mu)), draws(idata, "beta_class")])
    ok = np.column_stack([np.zeros(len(mu)), draws(idata, "beta_class_interval")])
    return {(c, t): expit(mu + bk[:, c] + t * (bt + ok[:, c])) for c in range(6) for t in (0, 1)}


cp = cell_probs(idata_C)
raw = d.groupby(["cls", "interval"]).response.mean()
rows = []
for (c, t), v in cp.items():
    lo, hi = hdi(v)
    rows.append({"класс": CLASS_RU[CLASS_ORDER[c]], "тест": "сразу" if t == 0 else "неделя",
                 "p, среднее": v.mean(), "94 % HDI": f"[{lo:.3f}, {hi:.3f}]", "сырая доля": raw[c, t],
                 "разность": v.mean() - raw[c, t]})
cell_table = pd.DataFrame(rows)
cell_table.round(3)
````

- Интервалы для обычных слов узкие, $\pm 0.04$–$0.05$: на ячейку приходится 800–1000 проб.
- Интервалы для ключевых слов в три-четыре раза шире, около $\pm 0.15$: ключевых слов около 200 на ячейку (по 5 на участника), а их эффект заметно различается между участниками и списками. Ключевое слово, которого не было: 0.44 [0.28, 0.59] сразу и 0.66 [0.51, 0.80] через неделю. Ключевое слово, которое сказал Миша: 0.62 [0.47, 0.77] и 0.61 [0.46, 0.76].

### 5.2 Контрасты: разности вероятностей

Научные вопросы — это разности ячеек. Для каждого значения из апостериорного распределения вычитаем одну вероятность из другой и получаем целое апостериорное распределение разности. Из него берём среднее, 94 % HDI и две вероятности, а по ним формулируем словесный вывод:

- $P(\text{контраст} > 0)$ — вероятность того, что эффект положителен;
- $P(\text{в ROPE})$ — вероятность того, что эффект **практически нулевой**: лежит в интервале $[-0.05, +0.05]$, то есть меньше 5 процентных пунктов. ROPE (region of practical equivalence) задаёт исследователь, а не данные. Здесь 5 п. п. — договорённость: разницу меньше 5 п. п. при 50 ответах участника никто не заметит.
- **Вывод:** эффект «практически ноль», если $P(\text{в ROPE}) \ge 0.9$; «положителен» или «отрицателен», если весь 94 % HDI лежит по одну сторону от нуля; иначе «не определён»: данные не позволяют ни утверждать эффект, ни исключить его.

Контрасты сгруппированы по вопросам из раздела «Эксперимент» плюс четыре **контрольных**: для обычных слов вербализация ничего менять не должна (ключевое слово ведь не касается этих слов), поэтому здесь ожидаем нуль. Если это не так, группы с вербализацией и без неё отличаются чем-то ещё.

````python
CONTRASTS = [   # (код, вопрос, описание, ячейка A, ячейка B); контраст = p_A − p_B; ячейка = (класс, момент теста)
    ("C1", "1. Криптомнезия", "ключевое (не звучало) − слово Миши, тест сразу", (2, 0), (0, 0)),
    ("C2", "1. Криптомнезия", "ключевое (не звучало) − слово Миши, через неделю", (2, 1), (0, 1)),
    ("C3", "1. Криптомнезия", "ключевое (не звучало) − своё слово, сразу", (2, 0), (1, 0)),
    ("C4", "1. Криптомнезия", "ключевое (не звучало) − своё слово, через неделю", (2, 1), (1, 1)),
    ("D1", "2. Неделя", "ключевое (не звучало): неделя − сразу", (2, 1), (2, 0)),
    ("D2", "2. Неделя", "ключевое (сказал Миша): неделя − сразу", (5, 1), (5, 0)),
    ("D3", "2. Неделя", "слово Миши: неделя − сразу", (0, 1), (0, 0)),
    ("D4", "2. Неделя", "своё слово: неделя − сразу", (1, 1), (1, 0)),
    ("V1", "3. Вербализация", "ключевое (сказал Миша) − слово Миши, сразу", (5, 0), (3, 0)),
    ("V2", "3. Вербализация", "ключевое (сказал Миша) − слово Миши, через неделю", (5, 1), (3, 1)),
    ("V3", "3. Вербализация", "ключевое: сказал Миша − не звучало, сразу", (5, 0), (2, 0)),
    ("V4", "3. Вербализация", "ключевое: сказал Миша − не звучало, через неделю", (5, 1), (2, 1)),
    ("K1", "Контроль", "слово Миши: с вербал. − без, сразу", (3, 0), (0, 0)),
    ("K2", "Контроль", "слово Миши: с вербал. − без, через неделю", (3, 1), (0, 1)),
    ("K3", "Контроль", "своё слово: с вербал. − без, сразу", (4, 0), (1, 0)),
    ("K4", "Контроль", "своё слово: с вербал. − без, через неделю", (4, 1), (1, 1)),
]
ROPE = 0.05


def contrast_draws(idata):
    """Выборки всех контрастов из апостериорного распределения: словарь код -> массив."""
    cps = cell_probs(idata)
    return {code: cps[a] - cps[b] for code, _, _, a, b in CONTRASTS}


def verdict(v, rope=ROPE):
    """Словесный вывод: практически ноль, если P(в ROPE) ≥ 0.9; иначе знак, если весь 94 % HDI по одну сторону от нуля."""
    lo, hi = hdi(v)
    if np.mean(np.abs(v) < rope) >= 0.9: return "практически ноль"
    if lo > 0: return "положителен"
    if hi < 0: return "отрицателен"
    return "не определён"


def contrast_table(idata, rope=ROPE):
    cd = contrast_draws(idata)
    rows = []
    for code, q, desc, a, b in CONTRASTS:
        v = cd[code]; lo, hi = hdi(v)
        rows.append({"код": code, "вопрос": q, "контраст": desc, "среднее": v.mean(), "94 % HDI": f"[{lo:+.3f}, {hi:+.3f}]",
                     "P(>0)": (v > 0).mean(), "P(в ROPE)": (np.abs(v) < rope).mean(), "вывод": verdict(v, rope)})
    return pd.DataFrame(rows).set_index("код")


tab_C = contrast_table(idata_C)
tab_C.round(3)
````

````python
def forest(cd_list, names, colors, title, codes=None, ncols=2):
    """Контрасты (среднее и 94 % HDI). Один анализ: общий график; несколько анализов: по панели на контраст."""
    codes = codes or [c[0] for c in CONTRASTS]
    desc = {c[0]: c[2] for c in CONTRASTS}
    if len(cd_list) == 1:
        fig, ax = plt.subplots(figsize=(10, 0.42 * len(codes) + 1.2))
        ax.axvspan(-ROPE, ROPE, color="0.85", alpha=0.7, label="ROPE ±0.05")
        ax.axvline(0, color="k", lw=0.8)
        for y, code in zip(np.arange(len(codes))[::-1], codes):
            v = cd_list[0][code]; lo, hi = hdi(v)
            ax.plot([lo, hi], [y, y], color=colors[0], lw=3, alpha=0.7, solid_capstyle="round")
            ax.plot(v.mean(), y, "o", color=colors[0], ms=5)
        ax.set_yticks(np.arange(len(codes))[::-1], [f"{c}  {desc[c]}" for c in codes], fontsize=8)
        ax.set_xlabel("разность вероятностей"); ax.set_title(title, fontsize=11); ax.legend(fontsize=8, loc="lower right")
    else:
        nrows = int(np.ceil(len(codes) / ncols))
        fig, axes = plt.subplots(nrows, ncols, figsize=(11, 1.0 + nrows * (0.28 * len(cd_list) + 0.7)), sharex=True)
        axes = np.atleast_1d(axes).ravel()
        for ax, code in zip(axes, codes):
            ax.axvspan(-ROPE, ROPE, color="0.85", alpha=0.7); ax.axvline(0, color="k", lw=0.8)
            for k, (cd, col) in enumerate(zip(cd_list, colors)):
                v = cd[code]; lo, hi = hdi(v); y = len(cd_list) - 1 - k
                ax.plot([lo, hi], [y, y], color=col, lw=3, alpha=0.8, solid_capstyle="round")
                ax.plot(v.mean(), y, "o", color=col, ms=5)
            ax.set_yticks(range(len(cd_list)), names[::-1], fontsize=8)
            ax.set_ylim(-0.6, len(cd_list) - 0.4)
            ax.set_title(f"{code}: {desc[code]}", fontsize=8, loc="left")
        for ax in axes[len(codes):]:
            ax.axis("off")
        for ax in axes[-ncols:]:
            ax.set_xlabel("разность вероятностей")
        fig.suptitle(title, fontsize=11, y=1.0)
    plt.tight_layout(); plt.show()


forest([contrast_draws(idata_C)], ["модель C"], ["tab:blue"], "Контрасты модели C: среднее и 94 % HDI")
````

**3. Вербализация.**
- Даже когда Миша произнёс ключевое слово, его присваивают намного чаще других слов Миши: +0.35 [0.21, 0.50] сразу и +0.19 [0.02, 0.34] через неделю (V1, V2).
- Вербализация **повышает** присвоение ключевого слова сразу: +0.18 [0.06, 0.30] (V3). Через неделю разность −0.05 [−0.16, +0.07]: направление и величина не определены (V4). Исходная гипотеза предполагала обратное: вербализация должна уменьшать присвоение.

**Контроль.** Своё слово с вербализацией против без: 0.003 [−0.05, +0.06] сразу (K3), $P(\text{в ROPE}) = 0.93$: практически ноль. Остальные три контроля лежат между −0.06 и −0.02: K1 и K2 не определены (интервал шире ROPE), а K4 ($-0.057\ [-0.114, -0.001]$) формально отрицателен, но его интервал едва не доходит до нуля. K1 и K4 лежат у границы нуля: при повторном запуске интервал K1 целиком лежит ниже нуля (раздел 5.6). Группы с вербализацией и без неё в обычных словах отличаются на 0–6 п. п. Это либо шум, либо различия между экспериментами. Следующий раздел покажет, какие из этих эффектов выдержат проверку составом экспериментов.

### 5.4 Как меняются выводы от модели к модели

Рассмотрим, какими были бы выводы при остановке на модели A или B. Для каждого контраста сравним среднее и ширину 94 % HDI.

````python
cmp_rows = {}
for name, idata in [("A", idata_A), ("B", idata_B), ("C", idata_C)]:
    cd = contrast_draws(idata)
    for code, *_ in CONTRASTS:
        lo, hi = hdi(cd[code])
        cmp_rows.setdefault(code, {})[f"{name}: среднее"] = cd[code].mean()
        cmp_rows[code][f"{name}: ширина HDI"] = hi - lo
cmp_tab = pd.DataFrame(cmp_rows).T
cmp_tab["ширина C / ширина A"] = cmp_tab["C: ширина HDI"] / cmp_tab["A: ширина HDI"]
cmp_tab.round(3)
````

- **Ширина интервала.** Для контрастов «ключевое слово против обычного слова» (C1–C4, V1, V2) в модели C она в 2.1–2.2 раза больше, чем в модели A. Для контрастов внутри ключевого класса (D1, D2, V3, V4) в 1.3–1.4 раза: список действует на обе сравниваемые ячейки одинаково, и часть неопределённости сокращается. Для обычных слов ширина почти не меняется (1.05–1.35).
- Контрасты, которые в моделях A и B выглядели бесспорными, в C стали менее уверенными. C1 («сразу») был $[0.09, 0.23]$, а стал $[-0.02, +0.30]$. Дело не в том, что C даёт другой ответ, а в том, что она учитывает, что ключевых слов пять и участники различаются.

### 5.5 Типичный участник или среднее по участникам?

Все вероятности выше посчитаны для **типичного участника в типичном списке**, у которого все случайные сдвиги равны нулю. Если участники и списки различаются сильно, средняя вероятность по всем участникам и спискам отличается от «типичной»: нелинейность логита сжимает вероятности к 0.5. Проверим, насколько. Для 400 значений апостериорного моделируем новых участников и списки (сдвиги из нормальных распределений с оценёнными разбросами) и усредняем вероятность.

````python
def population_average_probs(idata, n_draws=400, n_new=2000, seed=1):
    """Средняя по новым участникам и спискам вероятность «я» в 12 ячейках (без сдвига эксперимента): (n_draws, 12)."""
    r = np.random.default_rng(seed)
    idx = r.choice(len(draws(idata, "mu_alpha")), n_draws, replace=False)
    mu, bt = draws(idata, "mu_alpha")[idx], draws(idata, "beta_interval")[idx]
    bk = np.column_stack([np.zeros(n_draws), draws(idata, "beta_class")[idx]])
    ok = np.column_stack([np.zeros(n_draws), draws(idata, "beta_class_interval")[idx]])
    s = {k: draws(idata, k)[idx] for k in ["sigma_alpha", "sigma_list", "sigma_own", "sigma_key", "sigma_list_key"]}
    out = np.zeros((n_draws, 12))
    for n in range(n_draws):
        za, zl, zo, zk, zlk = (r.standard_normal(n_new) for _ in range(5))
        for c in range(6):
            for t in (0, 1):
                eta = mu[n] + bk[n, c] + t * (bt[n] + ok[n, c]) + s["sigma_alpha"][n] * za + s["sigma_list"][n] * zl
                if c in (1, 4): eta = eta + s["sigma_own"][n] * zo
                if c in (2, 5): eta = eta + s["sigma_key"][n] * zk + s["sigma_list_key"][n] * zlk
                out[n, c * 2 + t] = expit(eta).mean()
    return out, idx


pop, idx_used = population_average_probs(idata_C)
typ = np.column_stack([cp[(c, t)][idx_used] for c in range(6) for t in (0, 1)])
chk = pd.DataFrame({"ячейка": CELL_LABEL, "типичный участник": typ.mean(axis=0), "среднее по участникам и спискам": pop.mean(axis=0),
                    "разность": (pop - typ).mean(axis=0)})
print("Наибольшее расхождение по ячейкам:", f"{np.abs(chk['разность']).max():.3f}")
chk.set_index("ячейка").round(3)
````

**Что видим.** Для обычных слов разница между «типичным» и «средним» не превышает 0.003. Для ключевых слов она до 0.026 (наибольшая для ключевого слова без вербализации через неделю: 0.66 против 0.64), потому что их разброс по участникам и спискам большой. Для контрастов поправка до 0.04, при интервалах шириной около 0.3. Поэтому различием между «типичным» и «средним» можно пренебречь, но помнить о нём стоит.

### 5.6 Можно ли доверять расчёту: расхождения, ESS и повторный запуск

Сэмплер сообщил о расхождениях: 5 из 4000 в модели B и 1 из 4000 в модели C. Это немного, но расхождение — предупреждение: в этом месте апостериорного траектория могла быть посчитана неточно, а оценки могли сместиться. «Расхождений мало» этого не опровергает, поэтому проверим четырьмя способами:

1. **Где расхождения.** При каких значениях разбросов $\sigma$ они случаются.
2. **Повторный запуск** с другим seed и более мелкими шагами (`target_accept = 0.995`). Если расхождений нет, а оценки контрастов не изменились, подозрение снимается. Если изменились, основным результатам доверять нельзя.
3. **R-hat и ESS самих контрастов.** Мы интерпретируем контрасты, а не параметры, поэтому диагностики нужны и для них. Для границ 94 % интервала важен **ESS хвостов** (`ess_tail`), а не только ESS середины распределения (`ess_bulk`).
4. **Pareto-$k$ для LOO.** PSIS-оценка LOO надёжна, пока $k \le 0.7$ у всех проб; у проб с $k > 0.7$ она ненадёжна.

````python
def divergence_where(idata, label):
    """При каких значениях разбросов σ случаются расхождения: σ в шагах-расхождениях и во всех шагах."""
    div = np.asarray(idata.sample_stats["diverging"]).astype(bool)
    P = post(idata)
    rows = {}
    for s in ["sigma_alpha", "sigma_list", "sigma_exp", "sigma_own", "sigma_key", "sigma_list_key"]:
        if s in P:
            v = P[s].transpose("chain", "draw").values
            rows[s] = {"медиана по всем шагам": np.median(v), "максимум по всем шагам": v.max(),
                       "в шагах-расхождениях": ", ".join(f"{x:.2f}" for x in v[div])}
    print(f"{label}: расхождений {int(div.sum())}, по цепям {div.sum(axis=1).tolist()}")
    return pd.DataFrame(rows).T


display(divergence_where(idata_B, "Модель B").round(3))
display(divergence_where(idata_C, "Модель C").round(3))
````

**Что видим.** В модели B во всех пяти расхождениях $\sigma_{exp}$ заметно выше типичного: 0.14–0.50 при медиане 0.05 (в четырёх из пяти 0.39–0.50, максимум по всем шагам 0.80); в одном из них к тому же $\sigma_{list}$ почти равен нулю. В модели C единственное расхождение тоже при высоком $\sigma_{exp} = 0.34$. Это знакомые «трудные места» иерархических моделей: хвост большого разброса, который оценивается всего по четырём экспериментам (данных о нём почти нет), и область вблизи нуля, где апостериорное образует «воронку» (трудная для сэмплера геометрия). Из положения расхождений нельзя заключить, влияют ли они на оценки, поэтому переходим к повторному запуску.

````python
print(f"Повторный запуск: другой seed, target_accept = {TARGET_ROB}")
idata_B2 = sample(model_B(d), target_accept=TARGET_ROB, seed=SEED + 1, name="B, повтор")
idata_C2 = sample(model_C(d), target_accept=TARGET_ROB, seed=SEED + 1, name="C, повтор")

cd1, cd2 = contrast_draws(idata_C), contrast_draws(idata_C2)
rows = {}
for code_, *_ in CONTRASTS:
    (l1, h1), (l2, h2) = hdi(cd1[code_]), hdi(cd2[code_])
    rows[code_] = {"основной запуск": f"{cd1[code_].mean():+.3f} [{l1:+.3f}, {h1:+.3f}]",
                   "повторный запуск": f"{cd2[code_].mean():+.3f} [{l2:+.3f}, {h2:+.3f}]",
                   "|Δ среднего|": round(abs(cd1[code_].mean() - cd2[code_].mean()), 4),
                   "|Δ границы HDI|": round(max(abs(l1 - l2), abs(h1 - h2)), 3)}
rep_tab = pd.DataFrame(rows).T
display(rep_tab)
print(f"Наибольшее изменение среднего контраста: {rep_tab['|Δ среднего|'].astype(float).max():.4f}; "
      f"границы 94 % HDI: {rep_tab['|Δ границы HDI|'].astype(float).max():.3f}")
````

**Что видим.** При повторном запуске (seed `SEED + 1`, `target_accept = 0.995`) в модели C расхождений нет, а в модели B осталось 2 из 4000 (раньше 5). Средние контрастов модели C изменились не больше чем на 0.002, границы 94 % HDI — не больше чем на 0.014. Словесные выводы раздела 5 не меняются, кроме двух контролей на границе нуля: у K1 верхняя граница интервала была $+0.001$, стала $-0.002$; у K4 обе границы остались по одну сторону от нуля, но верхняя ($-0.001$) практически лежит на нуле. Поэтому K1 и K4 нельзя считать надёжными «отличиями от нуля»: это граничные случаи, и мы их так и читаем. Модель B нужна нам только как ступень к модели C (сравнение и проверка предсказаниями); итоговые результаты даёт C.

````python
import xarray as xr


def contrast_diag(idata):
    """R-hat и ESS (bulk и tail) самих контрастов: именно их мы интерпретируем."""
    cd = contrast_draws(idata)
    ds = xr.Dataset({k: (("chain", "draw"), np.asarray(v).reshape(CHAINS, -1)) for k, v in cd.items()})

    def vals(x):
        x = x.to_dataset() if hasattr(x, "to_dataset") else x
        return np.array([float(x[k]) for k in cd])
    return {"R-hat контрастов (max)": vals(az.rhat(ds)).max(),
            "ESS bulk контрастов (min)": vals(az.ess(ds, method="bulk")).min(),
            "ESS tail контрастов (min)": vals(az.ess(ds, method="tail")).min()}


def pareto_k_max(idata):
    """Наибольшее значение Pareto-k по всем пробам (нужен log_likelihood)."""
    return float(np.nanmax(np.asarray(az.loo(idata, pointwise=True).pareto_k)))


def diag_report(fits, contrasts=True):
    """Таблица диагностик для словаря {имя: idata}: расхождения, R-hat и ESS всех параметров, а также ESS контрастов."""
    rows = {}
    for name, idata in fits.items():
        rec = {k: v for k, v in DIAG_LOG[name].items() if k != "из"}
        if contrasts:
            rec.update(contrast_diag(idata))
        rows[name] = rec
    tab = pd.DataFrame(rows).T
    tab["расхождений"] = tab["расхождений"].astype(int)
    return tab


fits_main = {"A": idata_A, "B": idata_B, "C": idata_C, "B, повтор": idata_B2, "C, повтор": idata_C2}
diag_main = diag_report(fits_main)
diag_main["max Pareto-k"] = [pareto_k_max(fits_main[n]) if n in ("A", "B", "C") else np.nan for n in diag_main.index]
diag_main.round(3)
````

**Что видим.**

- **R-hat** не больше 1.008 у всех параметров (включая сдвиги $z$) и не больше 1.003 у контрастов.
- **ESS контрастов** модели C: середины распределения не меньше 1190, хвостов не меньше 1627 (при повторном запуске 1879 и 2167). Границы 94 % интервала опираются на хвосты, а их ESS достаточен. Наименьший ESS параметров — 785 у $\sigma_\alpha$ (модель C), 876 у модели B: разбросы оцениваются хуже всего, но для выводов о контрастах они нужны меньше всего.
- **Pareto-$k$.** Наибольшее значение: 0.05 (A), 0.26 (B), 0.51 (C). Ни у одной из 8350 проб $k$ не превышает 0.7, значит по этому критерию оценки LOO и разность около 50 единиц надёжны.

**Вывод.** Мы не нашли признаков, что 1–5 расхождений на 4000 шагов исказили оценки контрастов: повторный запуск с другим seed и более мелкими шагами даёт те же результаты, ESS достаточный. Это не доказательство отсутствия смещения в хвостах разбросов $\sigma$: в модели B расхождения при повторном запуске не исчезли совсем, поэтому выводов о значениях $\sigma_{list}$ и $\sigma_{exp}$ мы не делаем.

## 6. Устойчивость: можно ли доверять выводам?

Контрасты из раздела 5 верны для **этой модели с этими приорами и этими данными**. Проверим два вопроса, от которых выводы могут зависеть:

1. Меняются ли они, если приоры другие?
2. Меняются ли они, если убрать один из четырёх экспериментов? Это важно из-за схемы: часть ячеек измерена только в одном-двух экспериментах (раздел 1.4).

Для проверок пересчитываем модель C с более осторожными настройками сэмплера: `target_accept = 0.995` (`TARGET_ROB`) и цепями покороче (`DRAWS_ROB` и `TUNE_ROB` значений). Расхождения, R-hat и ESS каждого пересчёта собраны в таблицу в разделе 6.4. PyMC может предупредить, что для части параметров (в первую очередь разбросов $\sigma$) эффективная выборка на цепь меньше 100; для самих контрастов мы проверяем ESS отдельно, в 6.4.

### 6.1 Чувствительность к приорам

Сравним три набора приоров: базовые (как выше; в коде, таблице и на рисунке этого раздела они обозначены `лабораторные`), **узкие** (все ширины вдвое меньше: приор сильнее прижимает эффекты и разбросы к нулю) и **широкие** (все ширины вдвое больше). Если выводы одинаковы при всех трёх, данных достаточно, чтобы приор не имел значения.

````python
PRIORS = {
    "узкие": {k: v / 2 for k, v in PRIOR.items()},
    "лабораторные": PRIOR,
    "широкие": {k: v * 2 for k, v in PRIOR.items()},
}
print("узкие:", PRIORS["узкие"])
print("широкие:", PRIORS["широкие"])

fits_prior = {"лабораторные": idata_C}
for name in ["узкие", "широкие"]:
    print(f"\nПриоры «{name}»")
    fits_prior[name] = sample(model_C(d, PRIORS[name]), draws=DRAWS_ROB, tune=TUNE_ROB, target_accept=TARGET_ROB,
                              name=f"C, {name} приоры")

cd_prior = [contrast_draws(fits_prior[n]) for n in ["узкие", "лабораторные", "широкие"]]
forest(cd_prior, ["узкие приоры", "лабораторные", "широкие приоры"], ["tab:green", "tab:blue", "tab:red"],
       "Контрасты при трёх наборах приоров (модель C)", codes=["C1", "C2", "D1", "D2", "V1", "V3", "V4", "K3"])
pri_tab = pd.DataFrame({n: {c: cd[c].mean() for c, *_ in CONTRASTS} for n, cd in zip(["узкие", "лабораторные", "широкие"], cd_prior)})
pri_tab["наибольшее различие"] = pri_tab.max(axis=1) - pri_tab.min(axis=1)
pri_tab.round(3)
````

**Что видим.** Контрасты при трёх наборах приоров различаются не больше чем на 0.06 (наибольшее для V1: 0.32, 0.35, 0.37), чаще на 0.01–0.04. Узкие приоры слегка уменьшают эффекты, прижимая их к нулю, широкие слегка увеличивают, но знаки и вывод остаются теми же. Единственное заметное изменение: контраст C1 при узких приорах $+0.114\ [-0.001, +0.233]$ едва касается нуля, а при базовых и широких заметно включает его. Данных достаточно, чтобы приоры не меняли картину. Если бы они её меняли, пришлось бы сообщать результаты при разных приорах.

### 6.2 Различаются ли эксперименты?

Схема не полностью скрещена, поэтому различия между экспериментами могут выглядеть как эффекты вербализации или недели. Начнём с сырых долей. Для ключевых слов покажем долю ответов «я» **в каждом эксперименте и группе** (в таблице строки: эксперимент, столбцы: группа).

Интервалы широкие, потому что модель не уверена в разбросе участников и списков; при пяти ключевых словах и 20 людях в группе любая группа может оказаться в этом разбросе. Проверка по предсказаниям не отвергает модель, поэтому проверим устойчивость выводов напрямую.

### 6.3 Убираем эксперименты по одному

Пересчитаем модель C пять раз: без каждого из четырёх экспериментов по очереди, и только на исходном эксперименте Попова_д (80 человек, схема 2×2 с случайным распределением по группам; сдвиг эксперимента в этом варианте не нужен). **Вывод.** Когда выборки неоднородны, среднее по всем может отличаться от того, что видно в каждой из них. В Попова_д эффекты недели и вербализации на ключевом слове не обнаружены: оценки малы (D1 +0.07, V3 +0.06), но интервалы широки: D1 $[-0.11, +0.23]$, V3 $[-0.11, +0.25]$, то есть данные одной выборки не исключают положительных эффектов до +0.2 и небольших отрицательных. В объединённом анализе эти эффекты возникают из-за Попова_к.

## 7. Ассоциативная сила и ошибочное присвоение (LogDice)

Четвёртый вопрос: если криптомнезия — это «знакомость без контекста», то слова, сильнее связанные с ключевым, должны казаться знакомее и присваиваться чаще. Это проверяемое следствие гипотезы: сам механизм «знакомость → присвоение» по этим данным не наблюдается, мы видим только связь LogDice с ответами. **LogDice** — мера ассоциативной связи слова с ключевым словом списка, рассчитанная по большому русскому корпусу: чем она выше, тем чаще слова встречаются рядом. Для ключевых слов LogDice не определён, поэтому здесь работаем с **обычными словами** (7475 проб с известным LogDice; слово «мягкость» в корпусе не нашлось).

Сначала рассмотрим свойства предиктора.

````python
ORD = ["ordinary_partner_noverb", "ordinary_self_noverb", "ordinary_partner_verb", "ordinary_self_verb"]
dl = d[(d.is_drm == 0) & d.logdice.notna()].copy()
dl["ld"] = (dl.logdice - dl.logdice.mean()) / dl.logdice.std()                  # стандартизованный LogDice
dl["oc"] = pd.Categorical(dl.stimulus_class, categories=ORD).codes              # 4 обычных класса: 0..3
dl["word"] = pd.factorize(dl.list_num.astype(str) + "_" + dl.stimulus)[0]       # номер слова
dl["pos"] = (dl.word_num - dl.word_num.mean()) / dl.word_num.std()              # стандартизованный номер слова в списке
n_words = dl.word.nunique()
n_per_word = dl.groupby("word").participant_id.nunique()
print(f"Проб: {len(dl)}, разных слов: {n_words}; участников на слово (число участников: число слов): {n_per_word.value_counts().to_dict()}")
print("Слова, которые слышали не все участники:", dl[dl.word.isin(n_per_word[n_per_word < d.participant_id.nunique()].index)].stimulus.unique().tolist())
print(f"Корреляция LogDice и номера слова в списке: {dl.logdice.corr(dl.word_num, method='spearman'):.2f}")
share_self_first = d[d.word_num == 1].groupby("experiment").source_type.apply(lambda s: (s == "self").mean()).reindex(EXPERIMENTS)
print("Доля списков, которые начинает участник, по экспериментам:", share_self_first.round(2).to_dict())

fig, axes = plt.subplots(1, 2, figsize=(11, 3.4))
w = dl.drop_duplicates("word")
axes[0].hist(w.logdice, 15, alpha=0.75); axes[0].set_xlabel("LogDice слова"); axes[0].set_title(f"{n_words} разных слов", fontsize=10)
axes[1].boxplot([w[w.word_num == k].logdice for k in range(1, 11)], positions=range(1, 11))
axes[1].set_xlabel("номер слова в списке"); axes[1].set_ylabel("LogDice"); axes[1].set_title("Ассоциация по позициям", fontsize=10)
plt.tight_layout(); plt.show()
````

**Что видим.** Всего 49 разных слов. 45 из них слышали все 159 участников, а четыре — «корабль», «скала», «переплет» и «обед» — только 80 участников: это десятые слова списков, которые стоят в группах без вербализации; в группах с вербализацией на их месте звучит КД (так же устроено слово «мягкость», но для него LogDice нет). В среднем получается по 153 ответа на слово. LogDice лежит между 5.8 и 11.6. Слова в начале списка в среднем сильнее связаны с ключевым: корреляция LogDice с номером слова $-0.34$ (ранг Спирмена). В трёх экспериментах из четырёх слова в нечётных позициях читает участник, в чётных Миша; в Попова_к порядок чередования уравновешен.

### 7.1 Модель D

Берём ту же логику, что в модели C, но для обычных слов: четыре класса (слово Миши и своё слово, без вербализации и с ней) × момент теста, три случайных сдвига, случайный наклон «своё слово» по участникам. Добавляем LogDice:

$$\eta_i = (\text{как в модели C для обычных слов}) + \beta_{ld}\, l_i + \beta_{ld \times own}\, \text{own}_i\, l_i + \sigma_{word}\, z_{w[i]},$$

где $l_i$ — стандартизованный LogDice слова, $\text{own}_i = 1$, если слово произнёс участник, и $z_{w[i]}$ — сдвиг слова $w$. Наклон LogDice для слов Миши — $\beta_{ld}$, для своих слов — $\beta_{ld} + \beta_{ld \times own}$. Приор на обе добавки $N(0, 1)$: поправка на одно стандартное отклонение LogDice (1.2 единицы) может изменить шансы ответа «я» в $e^{\pm 1.9}$ = от 0.15 до 6.6 раза (94 % приора). Это слабый приор.

**Случайный сдвиг слова $\sigma_{word}$ — главное отличие от модели C.** LogDice принадлежит слову, а не пробе: у слова «волны» он один и тот же для всех 159 участников. Если считать пробы независимыми, получится, что мы знаем про LogDice больше, чем есть на самом деле: информации не столько, сколько 7475 проб, а столько, сколько различных слов (49). Сдвиг слова забирает на себя всё, что отличает слово от других, кроме LogDice. Сравним две версии: без и с этим сдвигом.

````python
def model_D(data, prior=PRIOR, word_effect=True, position=False):
    """Обычные слова: модель C для 4 классов + наклон LogDice (для слов Миши и своих) [+ сдвиг слова] [+ позиция в списке]."""
    g = grouping(data)
    own = data.oc.isin([1, 3]).values.astype(float)                           # слова, которые произнёс сам участник
    with pm.Model(coords={"class": ORD[1:]}) as model:
        eta = fixed_effects(data.oc.values, data.interval.values, prior) + random_shifts(g, prior)
        sigma_own = pm.HalfNormal("sigma_own", prior["s_slope"])
        z_own = pm.Normal("z_own", 0, 1, shape=g["n_j"])
        beta_ld = pm.Normal("beta_logdice", 0, 1)                             # наклон LogDice для слов Миши
        beta_ld_own = pm.Normal("beta_logdice_own", 0, 1)                     # на сколько наклон для своих слов отличается
        eta = eta + sigma_own * z_own[g["j"]] * own + (beta_ld + beta_ld_own * own) * data.ld.values
        if word_effect:
            sigma_word = pm.HalfNormal("sigma_word", 0.5)
            z_word = pm.Normal("z_word", 0, 1, shape=int(data.word.max()) + 1)
            eta = eta + sigma_word * z_word[data.word.values]
        if position:
            eta = eta + pm.Normal("beta_position", 0, 1) * data.pos.values
        pm.Deterministic("slope_partner", beta_ld)
        pm.Deterministic("slope_own", beta_ld + beta_ld_own)
        pm.Bernoulli("y", logit_p=eta, observed=data.response.values)
    return model


print("Модель D без сдвига слова")
idata_D0 = sample(model_D(dl, word_effect=False), draws=DRAWS_ROB, tune=TUNE_ROB, target_accept=TARGET_ROB, name="D, без сдвига слова")
print("Модель D со сдвигом слова")
idata_D1 = sample(model_D(dl, word_effect=True), draws=DRAWS_ROB, tune=TUNE_ROB, target_accept=TARGET_ROB, name="D, со сдвигом слова")
SLOPES = ["slope_partner", "slope_own"]
tabs = {}
for name, idata in [("без сдвига слова", idata_D0), ("со сдвигом слова", idata_D1)]:
    t_ = summarize(idata, SLOPES + (["sigma_word"] if "sigma_word" in idata.posterior else []))
    t_["P(>0)"] = [np.mean(draws(idata, v) > 0) if v in SLOPES else np.nan for v in t_.index]
    tabs[name] = t_
    print(f"\n{name}:")
    display(t_.round(3))
````

**Что видим**

- **Без сдвига слова** наклон для слов Миши $\beta_{ld} = 0.155\ [0.086, 0.226]$, $P(>0) = 1.00$, для своих слов $0.027\ [-0.039, +0.085]$.
- **Со сдвигом слова** наклон для слов Миши $0.139\ [0.014, 0.273]$, $P(>0) = 0.976$, для своих слов $-0.015\ [-0.149, +0.110]$. Среднее почти то же, **интервал вдвое шире** (стандартное отклонение 0.070 вместо 0.038). $\sigma_{word} = 0.38$: слова различаются между собой существенно, и не только из-за LogDice.
- **Как это читать.** Связь согласуется с гипотезой: чем сильнее слово связано с ключевым, тем чаще участник приписывает себе слово, которое на самом деле произнёс Миша. Но это связь, а не доказанный механизм «знакомость → присвоение»: LogDice — свойство слова, и у слов с высоким LogDice могут быть другие общие свойства (позиция в списке, частота, смысл). Для слов, которые сказал сам участник, оценка наклона близка к нулю, но интервал включает и небольшие положительные, и отрицательные значения: отсутствие связи это не доказывает. Размер эффекта небольшой: +1 SD LogDice (1.2 единицы) умножает шансы на $e^{0.14} = 1.15$, при вероятности около 0.31 это +3 п. п.

#### Картина: вероятность ответа «я» в зависимости от LogDice

Для ячейки «тест сразу, ключевое слово не звучало» нарисуем долю ответов «я» для каждого слова отдельно среди тех, кто его произносил сам, и среди тех, кто слышал его от Миши (точки; на точку приходится 12–28 ответов) и предсказания модели D для типичного участника с типичным словом: линия и 94 % полоса. Слова Миши и свои слова отдельно.

````python
grid = np.linspace(-2.2, 2.0, 60)
mu, bt = draws(idata_D1, "mu_alpha"), draws(idata_D1, "beta_interval")
bk = np.column_stack([np.zeros(len(mu)), draws(idata_D1, "beta_class")])
sl = {"partner": draws(idata_D1, "slope_partner"), "own": draws(idata_D1, "slope_own")}
cell_dl = dl[(dl.interval == 0) & (dl.verbalization == 0)]
word_rate = cell_dl.groupby(["word", "oc"]).agg(rate=("response", "mean"), ld=("ld", "first"), n=("response", "size")).reset_index()

fig, ax = plt.subplots(figsize=(8, 4.3))
for src, col, cls_ in [("partner", "tab:blue", 0), ("own", "tab:green", 1)]:
    p = expit((mu + bk[:, cls_])[:, None] + sl[src][:, None] * grid[None, :])
    lo, hi = np.quantile(p, [0.03, 0.97], axis=0)
    ax.fill_between(grid, lo, hi, color=col, alpha=0.2)
    ax.plot(grid, p.mean(axis=0), color=col, lw=2, label=("слова Миши" if src == "partner" else "свои слова") + ": модель D")
    sub = word_rate[word_rate.oc == (1 if src == "own" else 0)]
    ax.plot(sub.ld, sub.rate, "o", color=col, alpha=0.6, label=("слова Миши" if src == "partner" else "свои слова") + ": доля по слову")
ax.set_xlabel("LogDice (стандартизованный)"); ax.set_ylabel("вероятность ответа «я»")
ax.set_title("Тест сразу, ключевое слово не звучало", fontsize=10); ax.legend(fontsize=8, ncol=2, loc="lower left")
plt.tight_layout(); plt.show()
````

**Что видим.** Линия для слов Миши слегка растёт, линия для своих слов практически плоская и лежит гораздо выше. Точки по словам разбросаны сильно: каждая точка — доля по 12–28 ответам, а один ответ — это 0 или 1. Поэтому визуально эффект почти не различим, и вывод держится на модели, которая учитывает все 7475 ответов и разброс слов.

### 7.2 Не сводится ли LogDice к позиции слова в списке?

Список DRM устроен так, что слова, сильнее связанные с ключевым, стоят в начале (корреляция LogDice с номером слова около $-0.34$). Раньше стоящие слова вообще лучше запоминаются, и их источник легче вспомнить или спутать. Добавим номер слова в список как ещё один предиктор (стандартизованный) и оценим, что остаётся от эффекта LogDice.

````python
print("Модель D со сдвигом слова и позицией в списке")
idata_D2 = sample(model_D(dl, word_effect=True, position=True), draws=DRAWS_ROB, tune=TUNE_ROB, target_accept=TARGET_ROB,
                  name="D, со сдвигом слова и позицией")
t_pos = summarize(idata_D2, SLOPES + ["beta_position", "sigma_word"])
t_pos["P(>0)"] = [np.mean(draws(idata_D2, v) > 0) if v != "sigma_word" else np.nan for v in t_pos.index]
t_pos.round(3)
````

**Что видим.** После добавления позиции в списке наклон LogDice для слов Миши уменьшается: $0.09\ [-0.03, +0.23]$, $P(>0) = 0.91$. Коэффициент позиции $-0.17\ [-0.27, -0.05]$: слова в конце списка присваивают реже. Оценка LogDice **чувствительна к спецификации модели**: она зависит от того, учтена ли позиция слова. LogDice и позиция связаны ($-0.34$), и данные этих списков не позволяют их разделить. Поэтому мы не можем сказать, какая часть эффекта «принадлежит» LogDice, а какая позиции, и тем более утверждать, что позиция его *объясняет*. Вывод про механизм «ассоциация → присвоение» нужно делать осторожно: направление правдоподобно, но величина и надёжность зависят от модели. Развести эффекты можно только данными, где LogDice и позиция слова не связаны.

#### Диагностика моделей D

Ни в одной из трёх моделей D расхождений нет (`target_accept = 0.995`). R-hat не больше 1.014, ESS параметров не меньше 473 (середина) и 783 (хвосты).

### Чему мы научились

1. **Приоры.** Выборка из приоров (`pm.sample_prior_predictive`) показывает, какие исходы эксперимента модель допускает до данных. Приор, неинформативный на шкале логитов ($N(0, 10)$), сильно информативен на шкале вероятностей: он сосредотачивает вероятности у 0 и 1. Поэтому приоры и эффекты оцениваются на шкале вероятностей.
2. **Правдоподобие.** $\prod_i p_i^{y_i}(1 - p_i)^{1 - y_i}$, считается в логарифмах. Формулу можно проверить руками: она совпадает с PyMC до третьего знака. При 200 пробах приор почти не важен, при пяти — важен, и это объясняет, зачем нужна иерархическая модель.
3. **Проверять нужно то, что модель не подгоняла.** Доли ячеек модель воспроизводит автоматически. Разброс между участниками и различия между списками — нет, и именно они показали, что модель B неполна.
4. **Случайные эффекты — не только сдвиги.** Участники различаются точностью и склонностью к криптомнезии, списки — силой ключевого слова. Модель C учла это и **вдвое расширила интервалы** для вопросов про ключевое слово, не изменив оценок.
5. **Информации столько, сколько уровней у фактора.** О силе ключевых слов её даёт пять списков, об эффекте LogDice — 49 слов. Число проб (8350) здесь вводит в заблуждение.
6. **Сравнение моделей.** LOO показал, что модель C лучше предсказывает пропущенные пробы (когда остальные ответы тех же участников и списков известны) на 50 единиц, несмотря на 137 эффективных параметров. Для новых участников и новых списков это не проверялось.
7. **Нет доказательства нуля — нет данных.** ROPE отличает «эффект практически нулевой» от «данных мало»; когда интервал шире ROPE, правильный вывод — «направление и величина не определены».
8. **Устойчивость.** Выводы нужно проверять при других приорах и при удалении частей данных. Здесь это показало, что часть эффектов в объединённом анализе происходит из одной выборки.
9. **Диагностика.** Расхождения сэмплера не списывают на то, что их «мало»: их разбирают (где лежат, меняется ли результат при более мелких шагах). R-hat и ESS смотрят не только у параметров, но и у самих контрастов, а для границ интервала важен ESS хвостов.

### Ограничения

- **Порядок чередования в Попова_к** (разд. 6.2) и разнородность выборок мы не моделировали: данных не хватает, чтобы отделить порядок от эксперимента.
- Случайные наклоны независимы друг от друга; в полной модели их можно связать (LKJ-приор на корреляции). Поэтому о связи точности и склонности к криптомнезии мы ничего не утверждаем.
- LOO оценивает предсказание пропущенных проб при известных остальных ответах тех же участников и списков; для новых участников и списков оно не проверялось.
- Эксперименты не рандомизированы между собой, поэтому различия между ними нельзя отнести к одной причине.
